# Wireless Networking and Mobile Computing

### 1. Mobile Computing for Medical Technology

Mobile computing is transforming healthcare by enabling continuous monitoring and automated control of physiological processes. The human body naturally maintains internal stability through biological control systems. One example is how the pancreas regulates blood glucose. When this system fails, diabetes occurs, requiring external regulation.

Mobile devices, wearables, and connected sensors can assist these control processes. Smartphones and smartwatches offer additional computation through local processing, edge computing, and cloud services. They support real-time therapy adjustments, patient feedback, and remote monitoring, while keeping healthcare professionals in the loop.

### 2 Diabetes Mellitus

#### 2.1 The Human Body and the Pancreas

The pancreas (*Bauchspeicheldrüse*) produces two main hormones:

* **Insulin:** allows cells to absorb glucose from the blood, lowering blood sugar.
* **Glucagon:** stimulates the liver to release stored glucose, raising blood sugar.

A healthy pancreas keeps blood glucose within range.

### 2.2 Effect of Diabetes Mellitus 

Diabetes is a long-term condition that affects how the body converts food into energy.

* **Type 1:** the pancreas produces little or no insulin.
* **Type 2:** the pancreas does not produce enough insulin, or the body is resistant to it.

Patients with Type 1 and some with Type 2 diabetes inject insulin, and sometimes glucagon, to maintain safe glucose levels.
Stable blood glucose reduces risks such as kidney failure, blindness, and cardiovascular disease.

A typical insulin therapy includes:

* **Bolus insulin:** fast-acting, taken with meals.
* **Basal insulin:** slow-acting, maintaining background glucose control.
* **Glucagon:** used in emergencies for very low blood sugar.


### 1.3 From Manual to Automated Therapy

Traditionally, patients manually tested blood glucose several times per day, calculated required doses, and injected insulin. Modern systems automate much of this workflow:

1. **Continuous Glucose Monitor (CGM):** measures blood glucose in real time.
2. **Smartphone:** calculates insulin recommendations and displays data.
3. **Insulin Pump:** injects insulin subcutaneously under smartphone or controller supervision.


Most commercial systems are open-loop, meaning users confirm or modify recommended doses before injection.

Closed-loop systems, also called artificial pancreas, complete the feedback loop. They measure glucose, compute the insulin dose, and deliver it automatically.

### 1.4 The PID Controller in OpenAPS

OpenAPS (Open Artificial Pancreas System) is an open-source platform for automated insulin delivery using CGMs and pumps. Its control core, **oref0**, is based on a PID controller that keeps glucose near the target value.

<center>
  <img src="../../Admin/mobile/dana_lewis.png" width="30%">
  <img src="../../Admin/mobile/openaps.png" width="25%">
</center>
The controller continuously receives glucose readings from the CGM and calculates the deviation from the target.

* **P (Proportional):** reacts to the current deviation between measured and target blood glucose (BG). The larger the difference, the stronger the insulin correction. It acts immediately to counter sudden changes, but on its own may cause oscillation around the target.

* **I (Integral):** accumulates past deviations over time. If glucose has been consistently above or below the target, the integral term compensates by gradually increasing or reducing insulin delivery. This eliminates steady-state error and corrects slow drifts that the proportional part alone cannot fix.

* **D (Derivative):** looks at how quickly BG levels are changing. It predicts future trends and counteracts rapid rises or drops by adjusting the insulin rate preemptively. This damping action prevents overshoot and helps smooth out spikes or sudden swings that could cause hypoglycemia.

Together, the three terms balance responsiveness and stability. The proportional term provides fast correction, the integral ensures long-term accuracy, and the derivative keeps the response smooth and safe.


### 1.5 Toward Fully Automated Systems

Closed-loop artificial pancreas systems regulate both basal and bolus insulin continuously. The next development stage is dual-hormone control, using insulin and glucagon together. This enables faster correction of high or low glucose events.
