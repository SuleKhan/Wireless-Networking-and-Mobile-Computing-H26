### 1. Audio Theory
#### 1.1 Audio signals

Audio signals are mechanical pressure waves traveling through a medium such as air, water, or solids. They are not electromagnetic waves and cannot propagate in vacuum. When an object vibrates, it compresses and rarefies the surrounding medium, creating a time-varying pressure pattern that travels to the listener. Microphones convert these pressure variations into electrical signals, and loudspeakers convert electrical signals back into mechanical motion.

Audio signals have a measurable bandwidth expressed in Hertz. Human speech energy mainly lies between roughly 100 Hz and 1 kHz, although intelligibility extends higher. The typical human hearing range reaches to about 15 kHz, and younger listeners may hear up to around 20 kHz. Many animals operate at different frequency ranges; elephants produce and detect very low-frequency sounds around 1 to 20 Hz to communicate over kilometers, while whales use frequencies from about 30 Hz to several kilohertz underwater.



#### 1.2 Spectrogram example

A spectrogram displays how the frequency content of a signal changes over time. The horizontal axis is time, the vertical axis is frequency, and the color represents signal level. For audio, the level is often expressed in decibels relative to full scale (dBFS) when working in digital form. This format makes it easy to analyze formants, harmonics, and energy distribution in speech and music. A spoken word such as “hello” shows rising and falling energy at distinct frequency bands corresponding to vocal tract resonances.

#### 1.3 Steganography in audio

Steganography hides information inside an ordinary signal to avoid drawing attention to the fact that data is present. In audio, this typically means embedding information at levels or frequencies that are not perceptually noticeable. Unlike encryption, which protects content by making it unreadable without a key, steganography protects by hiding the existence of the message. The term originates from the Greek “steganos” meaning covered, and “graphein” meaning writing.

This is related but not identical to watermarking and cryptography:

| Concept       | Goal                              | Typical method                           | Visibility / detectability       |
| ------------- | --------------------------------- | ---------------------------------------- | -------------------------------- |
| Cryptography  | Prevent unauthorized reading      | Mathematical encryption                  | Message visible, content hidden  |
| Steganography | Hide the existence of a message   | Embed in a carrier (audio, images, text) | Communication hidden             |
| Watermarking  | Embed ownership or integrity info | Perceptually masked embedding            | Presence detectable, data robust |
