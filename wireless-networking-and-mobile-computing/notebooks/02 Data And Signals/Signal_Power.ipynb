#### 2.1 Signal power and distance

When a transmitter radiates power, the energy spreads outward in space. As distance increases, the same amount of energy is distributed over a larger spherical surface. Because the surface area of a sphere grows with $4 \pi r^2), the received signal power decreases with the square of the distance. This relationship is called the inverse-square law, and it sets the fundamental scaling for wireless range.

$$
P_{RX}(r) \propto \frac{1}{r^2}
$$
$$
F(r) = 4 \pi r^2
$$

A larger receiving antenna aperture captures more of the spreading wavefront.

$$
s(r) = s_0 \cdot \frac{A}{4 \pi r^2}
$$

Example
At one meter: $P_{RX} = 1\text{ mW}$
At three meters: $P_{RX} = \frac{1}{9}\text{ mW}$

#### 2.2 Signal power units (linear and logarithmic)

Wireless systems often use decibel (dB) units because they simplify multiplication and division into addition and subtraction. Power in milliwatts uses dBm and power in watts uses dBW.

$$
s[\text{dBm}] = 10 \log_{10}\left(\frac{s[\text{mW}]}{1\text{ mW}}\right)
$$



```python
import numpy as np
import matplotlib.pyplot as plt

# Data
s_mw = np.linspace(0.001, 10, 500)   # 1 µW to 10 mW
s_dbm = 10 * np.log10(s_mw / 1)      # reference = 1 mW

fig, (ax1) = plt.subplots(1, 1, figsize=(10, 5))

# Left: linear scale
ax1.grid(True, color="white", alpha=0.2)
ax1.plot(s_mw, s_dbm, linewidth=2)
ax1.set_xlabel("Power (mW)")
ax1.set_ylabel("Power (dBm)")
ax1.set_title("mW to dBm (linear axis)")
ax1.set_facecolor((0.2, 0.2, 0.3))
ax1.legend([r"$10\log_{10}(mW)$"], fancybox=True, framealpha=1, loc="upper left", facecolor=(0.6, 0.61, 0.8))

plt.show()
```

$$
s[\text{dBW}] = 10 \log_{10}\left(\frac{s[\text{W}]}{1\text{ W}}\right)
$$


```python
import numpy as np
import matplotlib.pyplot as plt

# Data
s_w = np.linspace(0.001, 10, 500)   # 1 mW to 10 W
s_dbw = 10 * np.log10(s_w / 1)      # reference = 1 W

fig, (ax1) = plt.subplots(1, 1, figsize=(12, 5))

# Left: linear
ax1.plot(s_w, s_dbw, linewidth=2)
ax1.grid(True, color="white", alpha=0.2)
ax1.set_xlabel("Power (W)")
ax1.set_ylabel("Power (dBW)")
ax1.set_title("W to dBW (linear axis)")
ax1.set_facecolor((0.2, 0.2, 0.3))
ax1.legend([r"$10\log_{10}(W)$"], fancybox=True, framealpha=1, loc="upper left", facecolor=(0.6, 0.61, 0.8))

plt.show()
```

Conversion
$$
s[\text{dBm}] = 10 \cdot \log_{10}(s[\text{W}]/1\text{W}) + 30
$$

$$
1\text{ dBW} = 30\text{ dBm}
$$

#### 2.3 Attenuation and amplification

Scaling power is easier on a dB scale. Multiplication in linear units becomes addition in dB.

Examples
- Doubling power: $+3\text{ dB}$
- Halving power: $-3\text{ dB}$
- Multiplying by 4: $+6\text{ dB}$
- Dividing by 10: $-10\text{ dB}$
- Dividing by 1000: $-30\text{ dB}$

If a signal drops by a factor of 0.001 in linear units, that is

$$
s_{RX}[\text{dBm}] = s_{TX}[\text{dBm}] - 30
$$

This model applies to amplifier gains, filter losses, cables, connectors, and free-space propagation effects.


#### 2.4 Free-space path loss

The path loss expresses how much signal power decreases as a wave travels through free space. It depends on wavelength (\lambda) and distance (r).

$$
PL(r) = 10 \log_{10}\left( \frac{\lambda}{4 \pi r^2} \right)
$$

Using wavelength (\lambda = c / f), higher frequencies have shorter wavelengths and higher path loss. In dB terms this becomes a subtraction from the transmit power.

$$
P_{RX}(r) = P_{TX}[dBm] + PL[r]
$$

Example
- Frequency: $(2.4\text{ GHz})$
- Transmit power: $10\text{ mW} = 10\text{ dBm}$
- Path loss at one meter: $-20.023\text{ dB}$
- Received power: $-10.023\text{ dBm}$

Higher frequency signals lose energy faster in free space due to smaller wavelength and higher spreading loss. This is one of the reasons mmWave systems require beamforming and dense deployment.



```python
import numpy as np
import matplotlib.pyplot as plt

freqs = [2.4e9, 5.4e9, 60e9]
labels = ['2.4 GHz','5.4 GHz','60 GHz']
c = 3e8

r = np.logspace(-1, 2, 400)  # 0.1m to 100m

for f, label in zip(freqs, labels):
    lam = c / f
    PL = 10 * np.log10((lam / (4*np.pi*r**2)))
    plt.plot(r, PL, label=label)

plt.xlabel("distance (m)")
plt.ylabel("path loss (dB)")
plt.title("Free-space path loss vs distance")
plt.xscale("log")
plt.legend()
plt.grid(True)
plt.show()
```

#### 2.6 Multipath and shadowing

Real-world propagation does not follow a perfect free-space model. Obstacles, reflections, scattering, and diffraction change the signal strength over time and location.

Multipath occurs when signals bounce off surfaces and combine at the receiver. Depending on phase, paths can reinforce or cancel each other. This leads to fading and fast power fluctuations over small distances.

Shadowing happens when objects block or partially block the signal, reducing power through absorption or scattering. Large structures like buildings or terrain create deep signal drops behind them.

#### 2.7 Atmospheric attenuation

At longer distances and higher frequencies, the atmosphere absorbs energy. Oxygen absorption peaks near 60 GHz and water vapor contributes most near 24 GHz and above 100 GHz. Rain, fog, and humidity affect microwave and millimeter-wave signals. Lower frequencies travel farther through the atmosphere with less absorption, which is why traditional cellular bands sit below 6 GHz.
