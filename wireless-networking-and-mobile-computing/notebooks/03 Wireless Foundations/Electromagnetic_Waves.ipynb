# Wireless Networking and Mobile Computing

### 1 Electromagnetic Waves
#### 1.1 Creating an Electromagnetic Wave

An electromagnetic wave begins when an electric current changes over time. If an alternating current flows through a conductor or an inductive coil, the current (I) and circuit resistance (R) relate through (I \cdot R), but more importantly the changing current produces a changing magnetic field (B). This changing magnetic field induces an electric field (E), and the changing electric field again produces a magnetic field. The fields build on each other and move outward, which allows the wave to leave the conductor and travel through space. A constant current or static charge does not radiate, only accelerating charges create electromagnetic waves.

#### 1.2 Integral forms of Maxwell's Equations

Gauss's Law (Electric)
$$\oint_{S} E \cdot dA = \frac{Q_{enc}}{\varepsilon_0}$$

Gauss's Law (Magnetism)
$$\oint_{S} B \cdot dA = 0$$

Faraday's Law
$$\oint_{C} E \cdot dl = -\frac{d}{dt} \int_{S} B \cdot dA$$

Ampère–Maxwell Law
$$\oint_{C} B \cdot dl = \mu_0 I_{enc} + \mu_0 \varepsilon_0 \frac{d}{dt} \int_{S} E \cdot dA$$

#### 1.3 Time-varying Electric and Magnetic Fields


When a charge accelerates, the electric and magnetic fields around it change in time. These changes do not stay at the source and instead travel outward as a wave. A changing electric field produces a magnetic field and a changing magnetic field produces an electric field, so the fields sustain each other as they move.

- Changing $E$ produces $B$; changing $B$ produces $E$.
- In free space the wave speed is $c = 1/\sqrt{\mu_0 \varepsilon_0}$.

#### 1.4 Radiation and Wave Propagation

At high enough frequency, the coupled electric and magnetic fields detach from the source and propagate through air or vacuum. Close to an antenna the fields are complex and energy is mostly stored near the conductor. Far away the fields form a traveling wave that carries energy outward. Simple textbook field drawings only show the basic idea. Real waves have more complex patterns that depend on frequency, antenna geometry, and the surrounding medium.

#### 1.5 Dipole Antenna

A dipole antenna consists of two conductive arms with an alternating current driving them at the center. As the current reverses direction, charges accelerate along the elements and create time-varying electric and magnetic fields. Close to the dipole the fields are reactive and mainly store energy, while farther away the fields form a radiating wave that carries energy into space. The dipole is a fundamental reference antenna because it has a simple structure, a clear radiation pattern, and predictable impedance, and many practical antennas are variations of the dipole idea.
