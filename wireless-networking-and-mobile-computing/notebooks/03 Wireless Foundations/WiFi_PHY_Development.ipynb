# Wireless Networking and Mobile Computing

### 1 IEEE 802.11 Reference Model

IEEE 802.11 separates functions into the physical layer and the medium access control layer. The physical layer handles modulation, coding, and transmission over the radio channel. The MAC layer handles channel access, framing, acknowledgments, association, and link security. In addition, the MAC Layer Management Entity coordinates management operations such as scanning and authentication. The Station Management Entity spans both layers and provides a control and configuration interface toward higher layers through Service Access Points.

**Layer roles**

- Layer 1: Physical transmission schemes. Modulation, coding, timing, synchronization, channel estimation, OFDM symbol generation

- Layer 2: MAC protocols and MAC Layer Management Entity. Channel access, CSMA CA, interframe spacing, retransmissions, association and authentication

- Layer 1 and 2: Station Management Entity. Configuration, scanning, timing alignment, power management, measurements, key management hooks

- Higher layers: Interfaced through Service Access Points (SAPs)

### 1 Early and Basic Physical Layers of IEEE 802.11

The first IEEE 802.11 standard (1997) provided simple wireless connectivity with data rates of 1 Mb/s and 2 Mb/s. These early systems were designed for robustness and operation in unlicensed spectrum, not high throughput. The physical layers relied on spread-spectrum techniques originally motivated by resilience to interference and regulatory requirements in the ISM band. Infrared was also included as an alternative, though it saw little practical adoption.

**Frequency Hopping Spread Spectrum (FHSS)**: 
FHSS rapidly switched between narrow frequency channels according to a pseudo-random hopping sequence shared between transmitter and receiver. Each hop used a narrowband waveform, and the rapid change in frequency provided robustness against narrowband interference. This approach spreads the signal energy over time across many channels, but at any instant the transmission occupies only one channel.

**Direct Sequence Spread Spectrum (DSSS)**: 
DSSS transmitted all the time on a single 2.4 GHz channel, but multiplied each bit by a fixed chip sequence to spread energy across a wider bandwidth. All stations used the same spreading code, so this was not CDMA. The wider bandwidth and correlation receiver improved interference resistance, but data rates were still limited to 1–2 Mb/s.

**Infrared (IR)**: 
A diffuse infrared physical layer was included, enabling short-range wireless links without radio. It relied on optical signals with direct line-of-sight or reflections. It was rarely adopted and soon phased out in practice.

#### 1.2 IEEE 802.11b (11 Mb/s DSSS Extension)

IEEE 802.11b (1999) extended DSSS operation in the 2.4 GHz band to support 5.5 Mb/s and 11 Mb/s. High-Rate DSSS (HR/DSSS) used the same channelization as the earlier standard but replaced the spreading modulation with Complementary Code Keying (CCK), which allowed each symbol to encode more information while maintaining correlation properties. 802.11b was the first standard widely known as Wi-Fi and the first with mass-market consumer adoption. It established the technology commercially and drove large-scale WLAN deployment.

#### 1.3 IEEE 802.11a (54 Mb/s OFDM)

IEEE 802.11a (1999) introduced OFDM in the 5 GHz band with 20 MHz channels and 52 subcarriers. It used the same MAC layer as legacy 802.11 but adopted a new physical layer optimized for multipath environments. OFDM brought a large increase in spectral efficiency and peak throughput, up to 54 Mb/s. Because it operated in 5 GHz and required new radio hardware, it was not backward compatible with 802.11b.

#### 1.4 IEEE 802.11g (54 Mb/s at 2.4 GHz)

IEEE 802.11g (2003) delivered OFDM at 2.4 GHz, combining the performance of 802.11a with backward compatibility to 802.11b. It could operate at 6–54 Mb/s using OFDM, but also supported legacy DSSS and CCK modes for interoperability with older devices. In mixed mode networks, legacy 802.11b devices reduced overall capacity because the medium access rules and signaling were constrained by compatibility requirements. 802.11g achieved broad adoption due to performance and compatibility, and its 2.4 GHz operation gave it greater range than 802.11a.

#### 1.5 IEEE 802.11n (MIMO and High-Throughput WLAN)

IEEE 802.11n (2007/2008) introduced MIMO and channel bonding, enabling up to 600 Mb/s in ideal configurations. Multiple antennas allowed spatial multiplexing, and optional bonding of two 20 MHz channels enabled 40 MHz transmission. 802.11n supported both the 2.4 GHz and 5 GHz bands and greatly improved throughput, link robustness, and range. Concepts introduced in 802.11n such as MIMO later evolved into multi-user MIMO in 802.11ac and beyond.

### 2 Specialized Wi-Fi PHY Extensions

####  2.1 IEEE 802.11ad (Very High Throughput at 60 GHz)

IEEE 802.11ad introduced operation in the 60 GHz millimeter-wave band, unlocking multi-gigabit data rates over short distances. At these frequencies, antennas are highly directional and links behave almost like optical beams: communication is strongest in line-of-sight, with limited penetration and fast signal drop-off when blocked. Strong atmospheric absorption (especially oxygen), higher free-space path loss, and narrow beams limit range but reduce interference and allow very high spatial reuse.

A defining aspect of 802.11ad is that it supports single-carrier and OFDM. OFDM brings spectral efficiency and resilience to reflections, while SC-PHY can be more energy-efficient and power-amplifier-friendly at mmWave frequencies. Common applications are high-throughput short-range links such as docking stations and in-room streaming. Outside consumer markets, directional 60–80 GHz radios are widely used for carrier backhaul and ultra-low-latency financial trading networks.

**Single-carrier mode (SC-PHY)**
- Wide envelope waveform with lower PAPR and simpler RF chain
- lower hardware complexity and lower PAPR at very wide bandwidths

#### 2.2 IEEE 802.11ah  (Wi-Fi HaLow for Sub-GHz Long-Range Wi-Fi)

IEEE 802.11ah, known as Wi-Fi HaLow, extends Wi-Fi into the sub-1 GHz ISM band. Operating around 900 MHz, it takes advantage of longer wavelengths that travel farther, bend around obstacles better, and penetrate walls more effectively than 2.4 GHz or 5 GHz signals. The sub-GHz band is divided into narrow channel blocks, and HaLow can use single channels for maximum range or bond multiple channels for higher throughput. It keeps an OFDM PHY structure, but scaled to narrower bandwidths to reduce power consumption and maintain long-range link budgets.

HaLow is built for situations where coverage, battery life, and device density matter more than raw throughput. It supports thousands of stations per access point and communication distances from hundreds of meters indoors to over a kilometer outdoors. This makes it suitable for smart buildings, logistics, agriculture, and industrial monitoring, where IoT devices must operate for years on battery power while staying connected across large areas. HaLow fills the gap between short-range IoT radios like Bluetooth or Zigbee and high-rate Wi-Fi, offering Wi-Fi compatibility with extended reach and reduced energy demand.
