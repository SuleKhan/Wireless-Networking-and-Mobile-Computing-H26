# Wireless Networking and Mobile Computing

### 1 OSI Reference Model

The ISO OSI model explains how communication systems are structured into layers. It enables interoperability and modular design so that physical transmission is separated from higher level logic. Wi Fi (IEEE 802.11) primarily defines Layer 1 (physical layer) and Layer 2 (data link layer). Understanding the full stack helps place wireless systems in context.

#### 1.1 OSI Layers

| Layer | Name         | Examples                                               |
| ----- | ------------ | ------------------------------------------------------ |
| 7     | Application  | HTTP, SMTP, FTP, SSH, WebRTC                           |
| 6     | Presentation | TLS, SSL, MIME, codecs                                 |
| 5     | Session      | gRPC sessions, RPC, NetBIOS                            |
| 4     | Transport    | TCP, UDP, QUIC                                         |
| 3     | Network      | IPv4, IPv6, ICMP, routing                              |
| 2     | Data Link    | Wi Fi MAC, Ethernet MAC, Wi Fi authentication, ARP     |
| 1     | Physical     | OFDM, QAM, modulation and coding, carrier sense timing |

#### 1.2 Layer 1: Physical Layer

Layer 1 defines how bits turn into electromagnetic signals. It specifies modulation, coding, timing, synchronization and RF behavior. Wi Fi defines OFDM subcarriers, channel widths, modulation and coding schemes, MIMO, beamforming, power control and clear channel assessment.

Key tasks
- bit transmission
- frequency channel selection
- carrier sensing and timing
- forward error correction
- synchronization and channel estimation
- power and bandwidth control

### 1.3 Layer 2: Link Layer

Layer 2 structures bits into frames and controls how devices share the medium. Wi Fi uses CSMA CA, acknowledgments, retransmissions and association procedures to manage access and reliability.

Key tasks
- MAC addressing and frame format
- listen before talk channel access
- acknowledgments and retransmission
- association and authentication in WLANs
- QoS queues and scheduling
