# Wireless Networking and Mobile Computing

### 1. MAC Evolution in IEEE 802.11

#### 1.1 Early MAC (1997)  (Distributed Access and Baseline Reliability)

The original 802.11 standard introduced a medium access scheme designed for unpredictable radio environments. There was no assumption that devices could detect collisions as on Ethernet, so the MAC relied on carrier sensing with collision avoidance and acknowledgments. The goal was to make wireless communication reliable even when signals fade or overlap.

This first generation MAC defined the core ideas that still exist in modern Wi-Fi. A station listens to the channel before sending, waits a contention window to avoid simultaneous access, and expects an acknowledgment for each frame. If no acknowledgment returns, the frame is retransmitted with a larger contention window. This approach allowed Wi-Fi to scale from very small home networks up to crowded public deployments, long before QoS or multi-user features existed.

Hidden node scenarios, interference, and legacy physical layers made collisions expensive. The MAC therefore introduced optional Request-to-Send and Clear-to-Send control messages and fragmentation of large frames. These mechanisms allowed the network to protect long data transfers and limit retransmission cost in hostile environments.

The design priorities at that time were simplicity, interoperability, and reliability. Performance came later once the basics were stable.

#### 1.2 802.11b Era  (First Real-World Stress Test)

When 802.11b arrived, Wi-Fi suddenly entered millions of homes and offices. The MAC did not change dramatically, but the real-world environment did. Many independent devices operated with limited coordination, often separated by walls or floors. Hidden nodes and collisions became common, and RTS CTS and fragmentation were often used by early access points to keep networks stable.

This period exposed the limitations of a purely distributed random access method in dense deployments and laid the groundwork for later QoS and scheduling extensions. Voice and video applications emerged soon after, highlighting the need for traffic prioritization beyond best-effort access.



#### 1.3 802.11e  (Quality of Service and Controlled Access)

Voice over IP, video conferencing, and streaming media created new expectations about latency and fairness. The MAC expanded to support differentiated service for multiple traffic types. Enhanced Distributed Channel Access introduced four access categories: background, best-effort, video, and voice. Each category received different contention parameters and inter-frame timings so that delay-sensitive streams could transmit sooner.

In addition to priority queues, 802.11e allowed stations to send multiple frames in one burst after gaining access, known as a transmission opportunity. This reduced overhead for back-to-back packets and made real-time flows smoother.

802.11e also defined a centralized scheduling mode. A hybrid coordinator at the access point could assign explicit transmission rights to stations. While this approach enabled more deterministic service for industrial or enterprise use, it was rarely used in practice. EDCA became the dominant QoS mechanism and remains fundamental to modern Wi-Fi behavior.



#### 1.4 802.11k k / r / v / w  (Enterprise Mobility and Management)

As Wi-Fi became a primary access technology in enterprise networks, roaming performance, measurement, and control became critical. Several amendments addressed these needs.

802.11k introduced measurement frameworks so stations could evaluate channel load and neighbor access points. This allowed smarter roaming and load balancing.

802.11r support enabled fast roaming by securely transferring session keys between access points. Instead of reauthenticating each time, a device could continue a session with minimal interruption.

802.11v gave the network a way to guide clients, suggesting better channels or access points and coordinating power saving more intelligently.

802.11w protected management frames from spoofing or disruption, preventing attacks that could forcibly disconnect clients.

Together, these amendments transformed Wi-Fi from a best-effort local technology into a managed infrastructure capable of seamless mobility.

#### 1.5 802.11n (High-Throughput MAC for MIMO)

As PHY rates climbed with MIMO and 40 MHz channels, traditional per-packet overhead became a major limiter. Even though the radio could move bits faster, the MAC still spent time on preambles, headers, and ACKs for every frame. 802.11n redesigned the MAC to keep up.

It introduced frame aggregation so many MSDUs or MPDUs could be sent in one burst with one preamble and one contention event. Block acknowledgment kept reliability while eliminating dozens of small ACK exchanges. Together, these reduced overhead from dominant to manageable.

Because MIMO and channel bonding can change transmission formats on the fly, the MAC also learned to negotiate spatial streams, coordinate channel width changes, and advertise capability sets dynamically 802.11n is where the MAC and PHY really started operating as a tightly coupled system rather than two loosely connected layers.

The result was the first generation of Wi-Fi that delivered real hundreds-of-Mbps throughput and set the foundation for later very-high-throughput standards.


#### 1.6 802.11s (Mesh Networking and Multi-Hop Coordination)

Wi-Fi originally assumed a simple model: stations talk to an access point and the access point forwards traffic. But many scenarios do not have central infrastructure, for example campus extensions, rural access, industrial facilities, public-safety networks, and disaster relief 802.11s addressed that by adding a self-organizing mesh mode

Stations automatically discover neighbors, establish peer links, and choose routing paths. The MAC handles path selection, link metrics, forwarding rules, and synchronization needed for reliable multi-hop communication. The default routing system (HWMP) blends proactive and on-demand techniques to balance latency and overhead.

802.11s made Wi-Fi able to form distributed, infrastructure-free topologies while still interoperating with standard stations and access points. It brought mesh networking into the mainline Wi-Fi stack instead of relying on proprietary hacks.

#### 1.7 Evolution Summary

| -- | -- | -- |
| Initial standard            | Reliable wireless packet delivery | CSMA CA with ACK and backoff, RTS CTS, fragmentation         |
| First widespread deployment | Operate in homes and offices      | Practical tuning, early congestion control                   |
| QoS era                     | Voice and video                   | EDCA, TXOP, Block ACK                                        |
| Enterprise mobility         | Seamless roaming                  | Measurement, roaming assist, management protection           |
| High-throughput Wi-Fi       | MIMO and high PHY rates           | Aggregation, improved scheduling, channel width coordination |
| Mesh networking             | Multi-hop connectivity            | Distributed routing and neighbor discovery                   |
