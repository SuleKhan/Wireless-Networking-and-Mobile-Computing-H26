# Wireless Networking and Mobile Computing


### 1. Fundamentals

#### 1.1 Electromagnetic fields and radiation

An antenna launches or receives electromagnetic waves by driving or sensing alternating currents that create changing electric and magnetic fields. Near the antenna the fields are reactive and tightly coupled to the structure, while in the far field they propagate and carry energy away. Antenna size, shape, and orientation relate directly to wavelength, which influences gain, coverage pattern, and polarization. The same structure that transmits can usually receive, although front-end electronics determine sensitivity and noise performance.

### 2. Large-scale communication and broadcast systems

#### 2.1 Tall broadcast / cellular tower

These towers elevate antennas to improve line-of-sight and reduce ground clutter, which increases coverage range and reduces multipath. They are typically passive radiators, but they operate with active transmit amplifiers and often include low-noise receive amplifiers at the base or in remote radio heads. Sectorization and MIMO arrays allow multiple beams to serve many users simultaneously. Primary use cases include wide-area voice/data coverage and public broadcast.

#### 2.2 Deep-space / radio telescope dish

Large parabolic reflectors collect extremely weak signals and focus them onto a sensitive low-noise receiver. These systems are receive-only in astronomy mode, and are highly active on the receive side with cryogenic LNAs and precise filtering to maximize signal-to-noise ratio. They operate at microwave and radio bands where atmospheric attenuation and pointing accuracy are critical. The emphasis is sensitivity and stability rather than transmit power.

### 3. Protected and specialized systems

#### 3.1 Military radome-covered antennas

Radomes conceal antenna type and pointing direction while protecting hardware from weather and debris. These systems can be active beam-steered arrays or mechanically steered dishes, depending on mission. They often support both transmit and receive and may include active phased-array elements for rapid beam control. Common uses include satellite communication, radar, and secure long-distance links under harsh conditions.

### 4. Consumer and commercial satellite reception

#### 4.1 Small satellite dish with feed cover

Compact dishes focus satellite downlink signals into a feedhorn and low-noise block (LNB), which is an active receive stage that amplifies and down-converts the signal before cable loss. They are usually receive-only in broadcast TV services, while broadband satellite modems handle both transmit and receive with separate electronics. Accurate pointing and clear sky view are essential due to narrow beamwidths. The plastic cap protects the feed and maintains performance in rain and snow. These are also used by Starlink for cell reception.

### 5. Terrestrial home reception and local links

#### 5.1 Home terrestrial TV antenna

Typically a Yagi or log-periodic array, this is a passive receiving antenna optimized for directional gain and moderate bandwidth. It is used to capture terrestrial broadcast signals from regional transmitters, and performance depends heavily on elevation and orientation. Amplifiers may be added, but only after ensuring clean signal reception, as strong-signal environments can overload active front-ends. Primary goal is reliable reception of moderate-power broadcast signals.

#### 5.2 5G antenna array on tower

Modern cellular panels often include active radio units and phased-array elements, especially at higher frequencies. Massive MIMO and beamforming enable directional energy steering for both transmit and receive, improving spectral efficiency and user capacity. These systems support multi-band operation, so multiple panels appear on the same mast. Backhaul capacity, cooling, and precise synchronization are part of the design challenge.

#### 5.3 Home Wi-Fi router with antennas

Wi-Fi antennas are usually small dipoles or internal patches, often paired for MIMO reception and transmission. The antennas themselves are passive, while the router includes active amplifiers for both transmit and low-noise receive paths. Placement and polarization matter more than raw power, due to reflections and building materials. Scenario focus is short-range, high-throughput connectivity in indoor multipath environments.

### 6. Active vs passive receiving

A passive antenna converts electromagnetic energy into voltage/current at its terminals and relies on the receiver’s front-end. An active receiving antenna integrates a low-noise amplifier at or near the element to boost weak signals before cable loss and interference. Passive systems are preferred where signals are strong or interference is high, while active systems help in long-cable or weak-signal scenarios, especially at microwave frequencies. Choice depends on noise environment, dynamic range needs, and installation constraints.
