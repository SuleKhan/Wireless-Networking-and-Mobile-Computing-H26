### 1. Accessing Scientific Literature

The [ETH Library](https://library.ethz.ch/en) provides institutional access to peer-reviewed databases such as:

* [IEEE Xplore](https://ieeexplore.ieee.org/Xplore/home.jsp)
* [ACM Digital Library](https://dl.acm.org)


When citing, always record **full bibliographic information**: author(s), title, journal/conference, year, DOI. A consistent academic format (e.g., IEEE, APA, or ACM style) is expected. 

Example (IEEE):

> [1] C. E. Shannon, “A Mathematical Theory of Communication,” *Bell System Technical Journal*, vol. 27, pp. 379–423, 1948.

For each paper you include:

* Cite the **original source**, not summaries or tertiary sources.
* Prefer **DOI links** for permanence and clarity.
* Avoid citing AI tool outputs directly; instead cite the original paper retrieved with their help.


### 2. Using AI Tools in Academic Research

AI tools can accelerate literature exploration. ETH Zurich’s [AI in Education](https://ethz.ch/en/the-eth-zurich/education/ai-in-education.html) guidelines recommend critical use and clear attribution.

* [Connected Papers](https://library.ethz.ch/en/locations-and-media/media-types/databases-standards-patents/connected-papers.html) is a literature research tool
* [Elicit](https://elicit.com/) excels in the efficient handling of literature
* [ResearchRabbit](https://www.researchrabbit.ai/) provides support in organizing and searching for literature
* [Scite](https://scite.ai/) supports the search and evaluation of scientific publications


### 3. Supplementary Sources

Wikipedia and educational pages can be valuable for:

* Clarifying terminology.
* Tracing early sources or standards.
* Identifying relevant reference material.

However, Wikipedia is a **tertiary source**, use it to *discover* primary and secondary literature, not as a citation in scientific writing.
