### 1. Introduction

Every phone measures the cellular network all the time, to choose the cell that serves it best. You can read these measurements, and with a few of them you can see in the real network what `Network_Cellular.ipynb` models: low and high frequency bands, the loss through building walls, neighbouring cells on the same frequency, and capacity shared by many users.

This notebook explains how to read the measurements on your phone, what they mean, and what to measure. The measurements take about one hour in total, and much of it fits on your way to and from ETH.

#### 1.1 Reading the Measurements

| Phone | How | What it shows |
|---|---|---|
| Android, any maker | the free app NetMonster or Network Cell Info Lite (they need location permission) | band, PCI, RSRP, RSRQ, SNR, channel number, cell ID, neighbour cells |

The values update every few seconds and may be out of date on an iPhone, so wait until they change before you note them. If your phone shows no numbers, use the network icon and speed tests, or borrow a classmate's phone for the readings and say so.

#### 1.2 What the Values Mean

- Band: the frequency band in use. A phone switches to a lower band where the signal is weak, because low frequencies reach further.
- PCI: the physical cell ID of the serving cell. Neighbouring cells use the same frequency and are told apart by their PCI, so a new PCI means a new cell: another sector or another site.
- RSRP: the reference signal received power, in dBm. It is the power of a single subcarrier, so it is much lower than the total received power. About -80 dBm is strong, -100 dBm fair, and -115 dBm weak.
- SINR or RSRQ: the quality of the signal, including interference from other cells.
- Channel number (EARFCN for LTE): the exact frequency within the band. An operator usually uses one channel per band.

When the icon shows 5G, the phone usually uses an LTE cell and a 5G cell at the same time, and the app shows both. Note the values of the LTE cell, and the 5G band if one is shown.

#### 1.3 Rules for Measuring

- Measure only in public places, such as squares, streets, public buildings, stations, public garages and trams, never at home or at work.
- Crop screenshots to the numbers. Do not include maps of your movements, location logs, or people.
- Do not change any settings beyond airplane mode, such as locking the phone to a band.

### 2. What to Measure

#### 2.1 Three Places

Measure on the same day at three public places: outdoors on a square or a street, inside a public building near its entrance or a window, and deep inside, for example in an underground station, a public garage, or the middle of a large building. At each place, note the network icon, the band, the PCI and the RSRP, and run one speed test.

#### 2.2 A Second Position

From your outdoor spot, walk a few hundred metres along a street, for example to the next tram stop. Note the PCI and the band again. If the PCI is still the same, walk on until it changes.

#### 2.3 Two Speed Tests

Run a speed test twice at the same public place: once at a busy time, for example in a station at 17:30, and once at a quiet time. Note the PCI, the band and the RSRP with each test, so you can tell whether both tests used the same cell.

### 3. Building Entry Loss

Walls and windows weaken the signal on its way into a building. If your outdoor and indoor readings come from the same cell (the same PCI and band), the difference between them is the building entry loss:

$$
L_{\text{building}} = \text{RSRP}_{\text{outdoor}} - \text{RSRP}_{\text{indoor}}.
$$

If the PCI or the band changed between the two places, the phone switched cells, and the difference is not the building entry loss.
