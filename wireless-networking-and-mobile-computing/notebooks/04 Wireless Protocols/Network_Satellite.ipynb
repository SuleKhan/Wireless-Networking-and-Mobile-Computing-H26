# Wireless Networking and Mobile Computing

### 1. Introduction

Finding accurate position and time has challenged humans for centuries. Early navigators used the Sun, Moon, and stars for orientation, but accuracy was limited and depended on weather and observation skill. The invention of mechanical clocks and sextants improved navigation, yet determining longitude remained difficult until the creation of precise marine chronometers in the 18th century.

Today, satellite navigation systems have transformed this problem. By measuring how long radio signals take to travel from satellites to receivers on Earth, we can determine location and time with meter to centimeter accuracy, anywhere, anytime, and in almost any conditions.

This exercise explores Global Navigation Satellite System (GNSS) technology, focusing on the Global Positioning System (GPS). You will learn how position and time are derived through trilateration, the frequencies used, and how GPS signals are generated, transmitted, and decoded.

### 2. Global Navigation Satellite System (GNSS)

#### 2.1 GNSS and GPS

The term GNSS (Global Navigation Satellite System) refers to satellite-based positioning, navigation, and timing systems that provide global coverage. It is not a single system but an umbrella term for all constellations that transmit signals allowing users on Earth to determine their position and precise time.

GPS (Global Positioning System) is one specific implementation of a GNSS. Developed and operated by the United States Department of Defense, GPS was the first fully operational satellite navigation system and remains the most widely used. Because of its early dominance, “GPS” is often used informally to mean satellite navigation in general, but it is only one part of the broader GNSS family.

Modern GNSS receivers can use signals from multiple constellations at once. For example, most smartphones and surveying instruments combine GPS and Galileo, and often also GLONASS or BeiDou. Using multiple constellations improves accuracy, reliability, and performance in difficult environments such as cities or mountains where satellite visibility is limited.

#### 2.2 Major GNSS systems

The main global GNSS constellations in operation today are:

* **GPS (United States):** The oldest and most widely adopted system, operational since the 1990s.
* **Galileo (European Union):** Europe’s independent GNSS, designed for high accuracy and civilian control.
* **GLONASS (Russia):** Developed by the Soviet Union and later modernized by Russia, offering full global coverage.
* **BeiDou (China):** China’s navigation system, expanded to global coverage in 2020, with additional regional services.

Together, these systems provide global positioning and timing for applications ranging from smartphone navigation to aviation, maritime operations, agriculture, and research.

#### 2.3 Orbital speeds, heights, and periods

The figure below shows the main satellite orbit types used for navigation.

A **geostationary orbit (GEO)** is about 35,786 km above Earth’s equator. Satellites here orbit once every 24 hours, matching Earth’s rotation, so they appear fixed in the sky. GEO orbits are ideal for communication, weather, and TV broadcasting, since ground antennas can stay pointed at one spot. One satellite covers about a third of the planet, but the distance causes noticeable signal delay (about 0.25 seconds one way).

A **low Earth orbit (LEO)** ranges from 300 to 2,000 km altitude. Satellites complete an orbit in roughly 90 minutes, so many are needed for continuous coverage. Systems like Starlink and OneWeb use large LEO constellations to provide global internet with very low latency (under 30 ms). The trade-off is high maintenance cost and increased risk of space debris.

The **Van Allen radiation belts** are regions of charged particles surrounding Earth. The inner belt extends from about 600 to 12,000 km, and the outer belt from 13,000 to 60,000 km. Satellites in medium Earth orbit (MEO), such as GPS at ~20,200 km, operate within these belts and require radiation protection. LEO satellites orbit below them, while GEO satellites stay above.

#### 2.4 Satellite lifetimes
Satellites typically operate for 5 to 20 years, depending on their orbit, design, and available fuel.
In **low Earth orbit (LEO)**, atmospheric drag slowly reduces altitude, so lifetimes range from a few months (below 400 km) to about 10 years (around 600 km). Propulsion systems can extend service life or control reentry.
In **medium Earth orbit (MEO)** and **geostationary orbit (GEO)**, where drag is negligible, satellites usually last 10 to 20 years, limited mainly by fuel, radiation exposure, and electronic aging. When decommissioned, LEO satellites burn up in the atmosphere, while GEO satellites are moved to a higher “graveyard orbit.”

**Starlink satellites** in LEO (~550 km) operate for about 5 years. They use ion thrusters for orbit maintenance and deorbiting, burning up within a year after shutdown. Without propulsion, they would reenter naturally in 5–7 years. Their short lifespan supports frequent upgrades in the large constellation.

#### 2.5 Satellite frequency bands

Satellite communication uses several standardized radio bands. Lower frequencies travel farther and handle bad weather better, but they offer limited bandwidth and need larger antennas. Higher frequencies deliver much higher data rates and smaller antennas, but require more power and suffer from atmospheric attenuation such as rain fade.

Lower frequency bands (L, S, C)

* better propagation
* longer range
* more reliable in bad weather
* lower bandwidth
* larger antennas

Higher frequency bands (X, Ku, Ka)

* higher bandwidth
* smaller antennas
* better for high-capacity point-to-point links
* higher power requirements
* more sensitive to rain and atmospheric losses

These trade-offs shape which band suits navigation, broadcast TV, broadband Internet, radar, or deep-space communication.

##### Spectrum governance

International Telecommunication Union (ITU)

* allocates global frequency bands for space services
* coordinates assignments, prevents interference, ensures fair global access
* defines regulatory frameworks satellites must follow

National regulators (FCC, Ofcom, etc.)

* assign domestic satellite spectrum, issue licenses
* enforce compliance with national law and ITU rules
* manage interference across commercial, government, and broadcast sectors

### 3. Trilateration

#### 3.1 Estimating distance

GNSS satellites send electromagnetic signals that travel at the speed of light,
$c = 300{,}000 \ \text{km/s}$.
The receiver estimates distance by measuring how long the signal takes to arrive:

$$
\text{Distance} = c \times (\text{Receive Time} - \text{Transmit Time})
$$

Each satellite broadcasts its precise position and the exact time of transmission. The receiver compares this with its own clock to find the signal’s travel time. The measured distance places the receiver somewhere on a sphere centered on that satellite. To find an exact position, signals from multiple satellites are needed.

Accuracy depends on signal bandwidth. Wider bandwidths give sharper correlation peaks, improving timing precision. The GPS L1 C/A code (~2 MHz) provides meter-level accuracy, while newer signals like GPS L5 or Galileo E5 (~20 MHz) allow sub-meter accuracy.

Advanced correction methods improve precision even further:

* **RTK (Real-Time Kinematic):** uses carrier-phase measurements and reference stations for centimeter accuracy.
* **PPP (Precise Point Positioning):** applies precise satellite orbit and clock data for similar accuracy without a local base station.


#### 3.2 Multi-lateration

To find a three-dimensional position (latitude, longitude, and altitude), a GNSS receiver must also correct for its internal clock error relative to GNSS time. This introduces four unknowns: three for position and one for clock bias. Each satellite measurement defines a sphere centered on the satellite with a radius equal to the measured distance (pseudorange).

If the receiver clock were perfectly synchronized, three satellites would be enough. In reality, it is not, so at least four satellites are needed to solve for both position and clock offset.

> Even a small timing error causes large position errors because signals travel at light speed. A 1 ms clock offset equals roughly 300 km of distance error. GNSS satellites use atomic clocks accurate to nanoseconds, but receivers typically use quartz oscillators that drift by microseconds. The receiver therefore estimates its clock offset together with its position, requiring four or more satellites.

Modern GNSS receivers usually track 7–12 satellites, often from multiple constellations such as GPS, Galileo, BeiDou, and GLONASS. Extra satellites increase accuracy, reduce noise sensitivity, and improve reliability in obstructed environments.
