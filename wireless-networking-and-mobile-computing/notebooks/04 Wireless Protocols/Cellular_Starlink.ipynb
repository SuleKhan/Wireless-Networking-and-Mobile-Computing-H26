# Wireless Networking and Mobile Computing

### 1. Starlink 
#### 1.1 Introduction

Starlink is a large low-earth-orbit (LEO) broadband system built by SpaceX. It operates thousands of satellites in ~500–550 km orbits, providing global Internet coverage with low latency. Unlike GEO systems, satellites move rapidly relative to the user, so constant handover, precise tracking and agile beamforming are essential.

Satellites have short lifetimes (around five years) due to atmospheric drag at this altitude, but this allows regular technology refresh. Starlink has already deployed over 7,000 spacecraft, more than all other active satellites combined.


#### 1.2 Gateway Ground Stations

Starlink still uses terrestrial gateways for backhaul, although inter-satellite laser links reduce dependence on ground relays. Gateways are larger dedicated stations with high-gain antennas and more power.

Two flavors exist:

**Small gateway dishes (<1 m)**

* Tx: Ku-band
* Rx: Ka-band
* Cost-optimized, many deployed regionally

#### 1.3 Capacity and Laser Inter-Sat Links

Each v2 satellite provides roughly 80 Gb/s capacity. Mixing v1 and v2 satellites yields a system throughput around 250 Tb/s.
Laser crosslinks allow satellites to pass traffic between one another without immediately downlinking to Earth, improving:

* global coverage, including oceans and polar regions
* latency (shorter space-hops vs long fiber routes)
* resilience against localized outages or gateway congestion

This architecture shifts Starlink toward an optical mesh network in space.


#### 1.4 Direct-to-Cell Service

Starlink has begun rolling out Direct-to-Cell support, enabling standard LTE-compatible phone connectivity directly through satellites. This uses:

* modified LTE waveforms and timing to accommodate long delay
* lower frequencies for better penetration and antenna efficiency
* massive MIMO-style processing onboard satellites

The service targets messaging first, then voice and data as capacity allows.
