# Wireless Networking and Mobile Computing

### 1. Bluetooth

#### 1.1 Introduction
Bluetooth is a short-range wireless technology used to connect nearby devices with low power consumption. The name comes from King Harald *Bluetooth* who unified Danish tribes, and the logo combines the runes for his initials ᚼ and ᛒ. The core idea was the same: unify devices that otherwise would not speak to each other. Operating in the 2.4 GHz ISM band, Bluetooth enables cable-like communication without physical connectors, making it foundational for headsets, input devices, IoT gadgets and wearables.

Classic Bluetooth and BLE share the same band but serve different roles. Classic focuses on continuous, higher-throughput links such as audio streaming and HID devices. BLE is tuned for extremely low duty cycles, letting sensors sleep most of the time and only wake briefly to send small packets.


#### 1.2. Protocol Principles

Bluetooth communication follows a structured stack. At the baseband, the link controller manages timing, addressing, packet framing and radio power. The link manager negotiates and maintains secure links. Above it, L2CAP adapts higher-level data to the radio schedule. The design prioritizes predictable timing and low energy rather than sustained high throughput.

Connections operate in small time slots of $625 \mu s$, using time division duplexing where devices alternate transmit and receive. A master coordinates communication and slaves only transmit when polled, which guarantees order and avoids collisions inside a piconet.

</center>

Bluetooth uses frequency hopping to improve reliability. The 2.4 GHz band is divided into narrow channels and devices rapidly hop between them in a pseudo-random pattern, typically up to about 1600 hops per second. This makes Bluetooth resistant to interference and able to coexist with Wi-Fi and other devices.

A single master with up to seven active devices forms a piconet. When devices participate in multiple piconets, scatternets emerge, enabling larger multi-hop structures. This flexibility is rarely used in consumer devices but matters in mesh-style deployments and research networks.

### 2. Bluetooth Low-Energy

Bluetooth Low Energy is designed around the idea that most devices on a body, in a room, or embedded in equipment do not need continuous streaming. They need to wake, speak briefly and efficiently, then disappear back into silence. BLE keeps the same 2.4 GHz band as classic Bluetooth but redesigns nearly everything else to reduce duty cycle and simplify the radio’s work. Instead of the “always connected” model, it favors advertisement events, quick handshakes and a predictable sleep-wake rhythm that allows coin-cell powered devices to last months or years.

#### 2.1 Radio Channels and Operation

BLE divides the 2.4 GHz band into narrow channels (1 MHz or 2 MHz depending on PHY), but unlike classic Bluetooth it uses only three primary advertising channels (37, 38 and 39). These are placed to avoid overlap with common Wi-Fi channels, reducing collision probability in crowded environments. Devices periodically broadcast short advertising packets on these channels; scanners listen, then optionally request more information or initiate a connection.

Once a connection forms, the devices hop over a larger set of data channels using a connection-specific channel map. This map can adapt, disabling channels that experience persistent interference. BLE timing is tightly scheduled. A device transmits only at specific anchor points negotiated during connection setup, which allows the radio to remain off until the exact moment needed. This is what keeps power consumption low: most of the world ticks while the radio sleeps.

BLE supports multiple PHY modes. The standard 1M PHY is the default. The 2M PHY increases throughput at the cost of range. The coded PHY (often associated with Bluetooth 5 “long range”) uses redundancy and coding to push signals further or through more challenging environments, which matters for industrial sensors and medical wearables in real environments rather than lab benches.

#### 2.2 Security and Pairing

BLE supports encryption, authentication and device identity protection as core functions. Pairing is the first handshake where devices exchange keys. Over time BLE has tightened its security model: older Legacy Pairing is still seen in simple devices, but modern implementations increasingly use LE Secure Connections, which offers stronger cryptography and mitigates man-in-the-middle risks.

There are several pairing methods, each chosen based on the device’s UI constraints and threat model:

* **Just Works**: no passkey, suitable for low-risk, friction-sensitive use like fitness gadgets. Vulnerable to MITM during pairing, but once bonded the link is encrypted.


* **Passkey Entry**: one device shows a six-digit code and the other enters it, preventing MITM during setup. Used when a screen or keypad exists (wearables, medical monitors).



* **NFC Assisted**: out-of-band key exchange via NFC; secure and simple but requires hardware support.
* **Pre-shared Keys**: used in specialized systems where devices ship with a secret or are provisioned ahead of time.

After pairing, bonding saves keys for future sessions so reconnecting is instant and secure. BLE also rotates advertiser addresses periodically (commonly ~15 minutes), which prevents long-term tracking of users walking around with tags, beacons or health patches.

#### 2.3 Applications and Behavior in Practice

BLE’s value shows up in simple, low-duty, real-world patterns: wake briefly, send tiny data, disappear again. The point is long battery life and quiet reliability, not streaming.

**Location beacons for child safety**
A toy or wrist tag quietly broadcasts a rotating BLE ID. Park-side receivers hidden around paths and entrances listen. If a child strays into an unexpected zone or stops appearing in scans, staff or parents get notified.

**Device-to-device mesh chat**
Phones relay short encrypted messages directly between people, useful in dense festivals or areas with overloaded networks. Messages hop across nearby devices until they reach the destination. No cell service needed, only short BLE bursts and cooperative relay.

**COVID-19 exposure notification**
Phones broadcast anonymous rolling identifiers and log nearby ones. If someone later reports infection, matching is done locally on each device. BLE enabled large-scale proximity detection without location tracking or central identity databases.

**Medical data transmission**
A prefilled injector detects skin contact, performs the injection and emits confirmation clicks. At the same moment, a tiny BLE radio logs the event and a nearby phones pick it up and records it to the cloud. Allows tracking of medicine intake of patients.
