# Wireless Networking and Mobile Computing

### 1. Delay Tolerant Networking (DTN)


#### 1.1 Introduction

Delay Tolerant Networking (DTN) was designed for communication environments where a stable, continuous connection cannot be guaranteed.
Instead of assuming an always-on path between sender and receiver, DTN accepts that links may be intermittent, delayed, or completely disconnected for long periods.
Messages are stored temporarily and forwarded later, following a store–carry–forward principle.

The concept originated with the Interplanetary Internet initiative, where space missions required reliable delivery despite minutes or hours of propagation delay and limited contact windows.
Today, the same principles apply to terrestrial networks such as remote villages, disaster zones, or mobile ad hoc networks where nodes may only meet occasionally.

#### 1.2 Core Principles

DTN sits as an overlay layer between the transport and application layers.
It introduces a new data unit called a bundle, which encapsulates application data along with timestamps and delivery information.
Each bundle is stored until a suitable contact becomes available, then forwarded to the next node along the path.

Key functions of the DTN layer:

* Persistent storage: messages are retained locally if the next hop is unreachable.
* Asynchronous forwarding: data moves opportunistically when links appear.
* Addressing and custody transfer: responsibility for delivery shifts hop by hop.
* Compatibility: supports existing transport layers like TCP or UDP through convergence adapters.

This makes DTN robust against disruptions that would normally break a conventional TCP/IP session.


#### 1.3 The Bundle Protocol

The Bundle Protocol (BP) is the core of the DTN architecture, standardized by the IETF (RFC 4838 and RFC 5050).
It defines how bundles are created, forwarded, and acknowledged.
Each bundle includes metadata such as source, destination, creation time, and lifetime.

When a node receives a bundle, it checks whether it can deliver it locally, forward it to another node, or store it until a route becomes available.
Intermediate nodes may hold data for seconds or even hours before forwarding.

Bundle Protocol versions now support:

* fragmentation for large messages,
* security extensions for authentication and integrity,
* administrative records for status reporting.

This architecture allows operation across deep-space links, low-bandwidth rural links, or intermittent mobile networks.

#### 1.4 Routing in DTN

Unlike continuous networks, DTN routing must account for time and contact probability rather than immediate link state.
Several strategies exist:

| Routing Type                    | Description                                            | Example Use                              |
| ------------------------------- | ------------------------------------------------------ | ---------------------------------------- |
| Direct Delivery             | Wait until source meets destination.                   | Sparse, predictable mobility.            |
| Epidemic Routing            | Flood messages to all contacts until delivered.        | Small networks, high redundancy allowed. |
| Spray and Wait              | Limit number of message copies to reduce overhead.     | Moderate density, mobile nodes.          |
| Probabilistic / PRoPHET     | Use encounter history to estimate delivery likelihood. | Human mobility or periodic contacts.     |
| Contact Graph Routing (CGR) | Plan routes based on known future schedules.           | Space missions and deterministic motion. |

Each node must balance delivery probability and resource cost — too much replication wastes bandwidth, too little risks data loss.
Hybrid and adaptive schemes combine these methods to optimize performance under changing conditions.

#### 1.5 Performance and Challenges

Real-world DTN deployments face several challenges:

* Variable contact duration: links may exist only for seconds.
* Storage constraints: limited buffer space for queued bundles.
* Energy usage: radio must wake up periodically to check for contacts.
* Security: bundles may travel across untrusted devices.
* Delay: delivery can take minutes to hours, depending on mobility patterns.

Despite these limitations, DTN offers a unique capability — guaranteed eventual delivery, even when no continuous route exists.
This makes it invaluable for disaster recovery, sensor data collection, humanitarian logistics, and off-grid communication.

Typical performance in small mobile DTN experiments shows:

* Delivery ratio above 95–99% for short text messages.
* Average delivery delay within tens of seconds for four-hop paths.
* Robust operation with intermittent connectivity and node mobility.


#### 1.6 Applications

Real-world uses of DTN include:

* Space communication networks: NASA’s deep-space missions rely on DTN to handle long propagation delays.
* Rural connectivity: projects in Africa and South America use DTN to move data between isolated villages via buses or drones.
* Disaster response: mobile devices form DTN clusters to exchange maps, status messages, and medical reports when the Internet fails.
* Humanitarian missions: organizations like the ICRC combine DTN and mesh protocols for resilient communication under strict privacy conditions.
