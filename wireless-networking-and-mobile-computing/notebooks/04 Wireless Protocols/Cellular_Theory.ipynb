# Wireless Networking and Mobile Computing

### 1. History of 3GPP Cellular Standards


#### 1.1 2G: GSM (Digital Voice and Messaging)

2G marked the shift from analog to digital cellular networks. GSM introduced secure digital voice calls, SMS messaging and SIM-based subscriber identity. Radio access was based on TDMA and narrowband channels. Voice quality improved and roaming became practical across regions. Data existed but was limited to low-bitrate services like GPRS and EDGE, suitable only for simple messaging and slow browsing.

#### 1.2 3G: UMTS (Mobile Internet Access)

3G expanded cellular beyond voice, targeting real Internet usage. UMTS introduced wideband CDMA to carry voice and packet data in parallel. This enabled web browsing, streaming media and early smartphone services. Latency was still noticeable and throughput modest compared to today, but this generation established mobile devices as Internet clients rather than just phones. HSPA and HSPA+ later boosted 3G data speeds significantly.

#### 1.3 4G: LTE (All-IP Broadband)

4G LTE replaced circuit-switched voice with a full IP core network. It introduced OFDM for the air interface, similar to Wi-Fi, enabling flexible spectrum and high peak rates. Latency dropped substantially and user experience became comparable to fixed broadband. LTE supported wide bandwidths, scalable channels and high mobility. Voice shifted to VoLTE rather than legacy switching. This generation truly delivered broadband everywhere.

#### 1.4 4G LTE-Advanced (Higher Capacity and Coverage)

LTE-Advanced extended LTE through carrier aggregation, advanced MIMO antenna systems and improved interference coordination. These features increased data rates, expanded coverage and made better use of fragmented spectrum. Cells could handle more users and more load, especially in dense urban areas. The technology laid foundational capabilities later reused and expanded in early 5G deployments.

#### 1.5 5G: New Radio and Service Expansion

5G builds on LTE design principles but expands them across a wider range of spectrum, including sub-6 GHz and millimeter-wave bands. It introduces more flexible OFDM numerology, extremely low latency targets and the ability to support very dense device deployments. 5G is not only about phones; it supports industrial automation, connected vehicles, extended-reality systems and massive IoT networks. Network functions move closer to users via edge compute, improving responsiveness and enabling real-time services.

### 2. Infrastructure and Network Architecture

Cellular networks divide the environment into many small coverage areas called cells. Each cell has its own antenna and radio equipment, typically running at low transmission power, often around ten watts or less. The overall coverage area is created by stitching these cells together so mobile devices can move between them without losing service. Each cell uses a specific range of frequencies, and neighboring cells coordinate frequency use to minimize interference.

A base station contains the radio transceiver and control logic needed to manage users, assign radio resources and handle mobility. This architecture allows continuous coverage, handover between sites and efficient spectrum use.

#### 2.1 Distributed and Virtualized Radio Functions

Modern systems separate the radio head and the baseband processing. The radio front-end (Remote Radio Head) is located near the antennas, while the Baseband Unit can be centralized or virtualized. This split enables pooling of processing resources and easier upgrades.

Key components and functions include:

- Multi-Access Edge Computing (MEC):
Compute resources placed close to users, allowing fast response for applications like industrial control or automotive systems.

- Baseband Unit (BBU):
Digital processing for radio signals, often pooled or virtualized to improve efficiency.

- Software Defined Networking (SDN):
Centralized control of network behavior through software rather than fixed hardware functions.

- Network Function Virtualization (NFV):
Running core network services on general compute platforms instead of dedicated telecom hardware.

- Self-Organizing Networks (SON):
Automated configuration and optimization for tasks like load balancing, interference management and handover tuning.

- Device-to-Device (D2D):
Direct communication between nearby devices without routing through the central core, improving efficiency and latency.

- Vehicle-to-Everything (V2X):
Connectivity for vehicles to infrastructure, pedestrians, networks and other vehicles for safety and automation.

- Remote Radio Head (RRH):
Radio equipment placed close to antennas, reducing cable loss and supporting flexible deployment.

#### 2.2 Cell Planning and Frequency Reuse

Cells are arranged and sized to cover different areas, from large macro layers down to micro, pico and indoor cells. Spectrum is reused across these cells: a frequency channel assigned to one location can be reused in another cell sufficiently far away to avoid harmful interference. In practice, a base station may operate with roughly $10$ to $50$ frequency channels depending on spectrum availability and planning.

Let

* $K$ = total number of frequency channels available in the network
* $N$ = number of cells in a frequency reuse cluster

Then the number of channels per cell is

$$
\frac{K}{N}.
$$

A **larger** cluster ($N$ higher) improves signal quality by reducing interference, but each cell receives fewer channels. A **smaller** cluster ($N$ lower) increases per-cell capacity through more aggressive reuse, but requires more advanced interference control and planning.

This reuse principle is the foundation of cellular network scaling and leads directly to later techniques such as cell splitting, sectorization and dense small-cell deployments in high-traffic areas.

#### 2.3 Scaling Capacity and Densification

Network capacity increases through several strategies:

* adding more frequency spectrum
* increasing the number of cells in the same area
* deploying smaller cell sizes, including microcells down to room-level
* sectorizing a cell into three to six directional antenna sectors
* dynamic frequency borrowing from neighboring cells or operators
* splitting large cells with micro-cells to handle user growth

### 3. LTE and 5G Radio Interface

The transition from 3G to 4G marked a major shift in radio technology. LTE introduced Orthogonal Frequency Division Multiplexing (OFDM) as its core air interface, similar to Wi-Fi. Instead of spread spectrum and CDMA techniques used in earlier systems, LTE uses many closely-spaced orthogonal subcarriers to transmit data efficiently.

This change allowed more flexible channel bandwidths, much lower latency, and efficient support for bursty packet-based traffic. LTE supports both Frequency Division Duplex (FDD) and Time Division Duplex (TDD) and scales across a wide range of spectrum allocations. Voice moved from dedicated circuit-switched handling to Voice over IP.

5G New Radio continues and extends this approach. It uses scalable OFDM numerology, supports much wider bandwidths and includes operation in both licensed and unlicensed spectrum. While the principle remains similar to LTE, 5G adds finer time granularity, broader frequency ranges (including millimeter-wave) and improved scheduling to support ultra-low latency and very high capacity when needed.

### 4. 3GPP Applications and Performance Requirements

Cellular networks support diverse applications, each with different performance expectations. Some services demand very high data rates, while others require low latency, reliability or dense device support. Below is a structured view of the market and QoS needs as provided.

| Market              | Mobile Application                                  | QoS Requirement                                         |
| ------------------- | --------------------------------------------------- | ------------------------------------------------------- |
| Mobile Broadband    | fast broadband, cloud storage                       | very high throughput (up to about 10 Gb/s), low latency |
| VR/AR/Hologram      | live content, design tools, gaming                  | high throughput, low latency                            |
| Smart City          | IoT, cloud processing, traffic systems              | low latency                                             |
| Automotive          | autonomous driving, massive sensing, edge computing | low latency                                             |
| Smart Manufacturing | edge computing, logistics, cameras                  | high throughput, low latency                            |
| Drones              | network-assisted operation                          | low latency                                             |
| Wearables           | body cameras, safety systems                        | low latency                                             |
| eHealth             | remote monitoring, diagnostics                      | low latency                                             |
| Retail / Payment    | interactive displays, smart kiosks                  | high throughput                                         |
| Energy              | smart grid intelligence, smart meters               | high throughput                                         |

These service areas demonstrate why modern cellular systems are designed for flexibility. Some users need extremely fast broadband, others need deterministic low-latency communication, and many applications consist of very small data packets from large numbers of devices.

5G does not prioritize one category over the others. It integrates them, allowing one network to support everything from high-rate mobile video to industrial sensing and real-time control.

### 5. 5G New Radio (NR) Air Interface

5G New Radio extends the OFDM approach introduced in LTE, but makes it far more flexible across frequencies, bandwidths and deployment types. The design supports everything from wide-area macro coverage to dense indoor hotspots and high-capacity millimeter-wave links.

5G uses scalable subcarrier spacing, allowing the same waveform family to operate efficiently at low and high frequencies. Lower frequencies below a few GHz use narrower subcarrier spacing for long-range coverage, while millimeter-wave bands use wider spacing and wide channels to support high data rates and short latency.

Across low-band, mid-band and millimeter-wave, NR supports:

* flexible bandwidths (from a few MHz to hundreds)
* different numerologies for propagation conditions
* time-division operation in many bands
* low latency support with shorter slot durations
* scalable beamforming and MIMO for range and capacity

#### 5.1 Scalable and Slot-Based Framework

5G frames and slots are structured for forward compatibility. Instead of one fixed timing, the slot structure scales based on subcarrier spacing. This allows shorter transmission time intervals for low-latency applications and continues to support scheduled broadband traffic.


The core points are:

* scalable OFDM numerology
* self-contained slot structure (data + control inside the same slot)
* low-latency scheduling support
* future-proof timing structure for new services

Self-contained slots reduce round-trip time, enabling low-latency control and fast uplink-downlink switching.


#### 5.2 Advanced Channel Coding

5G replaces LTE’s turbo codes with LDPC (for data channels) and Polar codes (for control channels). These coding schemes support large code blocks efficiently and reduce decoding latency.

Benefits include:

* high coding gain and efficiency
* reliable large data blocks
* improved control channel reliability
* reduced processing delay

This combination improves throughput and robustness for broadband, machine-type and low-latency traffic.

#### 5.3 Massive MIMO and Beamforming

5G deployments widely use massive MIMO to increase capacity and coverage. Arrays with many antenna elements steer beams toward users and track their movement. This improves link quality, especially at mid-band and mmWave frequencies where directional gain matters.

Key aspects:

* multi-user MIMO with many antenna elements
* spatial multiplexing for higher sector capacity
* beamforming for range and penetration
* beam-tracking for mobile devices

Massive MIMO is essential for urban high-density capacity growth.


#### 5.4 Mobile mmWave

In high-frequency spectrum (e.g., 28 GHz), 5G delivers very wide channels and extreme throughput. Directional beams overcome higher path loss and support fast short-range links.

Typical mmWave traits:

* channel widths in the hundreds of MHz
* beamforming and beam-tracking mandatory
* strong performance in dense hotspots and fixed wireless
* extremely high capacity with limited coverage range

mmWave complements low/mid band 5G rather than replacing them.



#### 5.5 Device Capability Scaling (RedCap Example)

5G supports reduced-capability (RedCap) devices for wearables, industrial sensors and simpler IoT products. These devices use narrower bandwidth, fewer antennas and simpler RF components.

### 6. Toward 6G and Reconfigurable Intelligent Surfaces

6G research looks beyond conventional base stations and antennas. One key direction is Reconfigurable Intelligent Surfaces (RIS). These are engineered surfaces with many tiny controllable elements that can reflect, steer or focus radio waves. Instead of signals bouncing unpredictably from walls or objects, RIS makes the environment itself part of the network.

RIS promises:

* controllable reflections instead of random multipath
* coverage extension in difficult indoor or urban zones
* lower power requirements than active repeaters
* beam shaping without adding full radio chains

The idea is not to transmit more power, but to shape propagation paths so users in shadowed or dense environments still receive strong links. Surfaces may be integrated into walls, building facades, ceilings or street furniture.

Visible-Light Communication (VLC) is also studied alongside RIS. VLC uses modulated light from LEDs for short-range data, offering high bandwidth in clean indoor environments. It does not replace radio but may complement it in high-capacity indoor networks or specialized environments.

WRC (World Radiocommunication Conference) decisions will determine future spectrum for beyond-5G systems, including possible low-band, mid-band and terahertz allocations.

### 7. 3GPP Release 18 and 5G-Advanced

3GPP Release 18 (2024) marks the start of 5G-Advanced. It builds on existing 5G deployments but focuses on stronger efficiency, intelligence and flexibility. The goal is to refine the platform before 6G arrives, not replace it.

Release 18 lays groundwork for:

* highly automated networks
* richer real-time industrial services
* more energy-efficient radio operation
* extended XR and multimedia capabilities

Recent additions published in the months following the main release expand support for positioning accuracy, low-power IoT profiles and improved multi-antenna performance.
