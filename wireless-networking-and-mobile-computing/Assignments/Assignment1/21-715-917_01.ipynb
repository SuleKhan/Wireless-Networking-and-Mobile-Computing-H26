<div class="alert alert-info">

4) The auto-correlation of the random array has a peak at $x=0$, a big gap to the constant auto-correlation. Explain why this peak shows up and what it represents.

<div class="alert alert-info">

(M) Answer: *(double click to edit)*

At x=0, there is no lag, so each element is multiplied by itself, giving squared values that are all positive and add up. This creates a peak, which represents the total energy (sum of squared values) of the random array. At other lags, different random values are multiplied, causing positive and negative products to largely cancel out, resulting in much smaller values.

### Step 3: Regulation (1 point)

1) In the introduction of this class, we talk about the concept of radio regulation. Explain the difference between radio regulation and standardization.

<div class="alert alert-info">

(M) Answer: *(double click to edit)*

Radio regulation is a legal and policy matter: authorities decide which frequencies may be used, what power limits apply, how spectrum is licensed, and how interference is managed. Standardization is a technical matter: industry and standards bodies agree on common interfaces, protocols, and methods so devices from different manufacturers can interoperate.

In short, regulation governs access and use of the spectrum, while standardization defines how systems communicate and operate consistently.

2) Explain why not everything in a radio system is standardized, and some parts like algorithms remain trade secrets and proprietary to individual parties.

<div class="alert alert-info">

(S) Answer: *(double click to edit)*

Not everything is standardized because standards only cover interoperability, not competitive differentiation. Proprietary algorithms, signal-processing methods, and implementation details give companies a technological edge, help them attract customers, and protect intellectual property from imitation or misuse.
