7. Discuss how the frequency bandwidth of the signal changes when you increase the bit rate of the data. What does this mean for spectrum regulation?

<div class="alert alert-info">

(M) Answer: As bit rate increases, the signal changes more quickly in time, so its spectrum spreads out over a wider frequency range around the carrier. In other words, higher data rates require more bandwidth. This matters for spectrum regulation because regulators limit how much bandwidth each transmission can occupy, so systems must be designed to fit within allocated channels and avoid interference with other users.
