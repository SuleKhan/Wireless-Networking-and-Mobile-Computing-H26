2) What is the difference between `Jemula++` and `Jemula802++`?

<div class="alert alert-info">

(M) Answer: Difference between Jemula++ and Jemula802++: Jemula++ (simulator/kernel) is a generic discrete-event simulation framework that knows nothing about networking. It supplies the event scheduler and event list, simulated time and timers, random variables, statistics evaluation and plotting, so it could drive any event-based simulation. Jemula802++ (simulator/engine) is a domain-specific simulator built on top of that kernel. It models IEEE 802 wireless networks, mainly 802.11 WLAN, by adding the PHY and protocol layers, traffic generators, XML scenario handling, KML/KMZ output and a web console, with its components reusing the kernel's event, handler and statistics machinery.

5) What determines the simulation effort, the amount of computational power required to simulate a certain duration?

<div class="alert alert-info">

(M) Answer: In Jemula, computational effort is mainly set by the number of discrete events that must be processed, rather than simulated duration alone. A longer run usually means more events; more stations and higher offered traffic create additional packet transmissions, backoffs, collisions, and retransmissions. Mobility also adds cost because position updates trigger radio and attenuation recalculations, while more detailed protocol models and statistics add work. Therefore, equal-duration scenarios can take very different wall-clock time depending on event density and model complexity.

6) How do you decide if a simulation ran long enough?

<div class="alert alert-info">

(S) Answer: Run until the metric of interest has passed its warm-up transient and its rolling mean and confidence interval stabilize to acceptable precision. Confirm across independent seeds; a long simulated duration alone is insufficient if results still trend or uncertainty remains high.
