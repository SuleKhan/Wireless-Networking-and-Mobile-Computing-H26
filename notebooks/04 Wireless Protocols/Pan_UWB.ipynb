# Wireless Networking and Mobile Computing

### 1. Ultra-Wideband (UWB)

Ultra-Wideband spreads a very low power signal across a large amount of spectrum, rather than occupying a narrow channel like Wi-Fi or BLE. It sends extremely short pulses or wide multi-carrier signals and keeps power low enough that it sits below the noise floor of other services. Regulators define tight emission limits, so UWB operates quietly alongside other radio systems. This underlay approach allows it to coexist rather than compete for a single slice of spectrum.

The practical value of UWB comes from timing precision. It measures how long a signal needs to travel between devices, not just its signal strength. That makes UWB very good at accurate distance and angle measurement indoors where reflections usually cause problems for other radios. Typical systems reach centimeter-level accuracy, which is why UWB is used in phone-to-tag location, secure keyless entry, robotics positioning and indoor navigation.

Commercial use focuses mainly between roughly 3 and 10 GHz. Real deployments often operate around 6 to 8 GHz, following FCC and ETSI emission masks. The radio sends short pulses, listens for returns and reconstructs distance from precise timing. Power use can be kept low because the radio is active for short bursts and spends most time idle.

#### 1.1 UWB PHY Variants

Two approaches exist historically: direct-sequence UWB and multiband OFDM. Both occupy very wide bandwidth, but one does it with short spread pulses and the other splits the band into wide OFDM sub-bands. Modern commercial systems follow 802.15.4a and 802.15.4z, which are based on pulse techniques with additional security and ranging capabilities.

| Feature          | DS-UWB                                       | MB-OFDM                                       |
| ---------------- | -------------------------------------------- | --------------------------------------------- |
| PHY technique    | Single-carrier direct sequence spreading     | Multiband OFDM                                |
| Frequency use    | Two main bands: 3.1–4.85 GHz and 6.2–9.7 GHz | Four band groups, each about 528 MHz          |
| Data rates       | 28, 55, 110, 220, 500, 660, 1000, 1320 Mb/s  | 53.3, 80, 106.7, 160, 200, 320, 400, 480 Mb/s |
| Modulation       | BPSK                                         | QPSK                                          |
| Coding           | Convolutional with interleaving              | Convolutional with interleaving               |
| Logical channels | 12                                           | 18 via multiplexing                           |

While the table lists high bit rates, commercial UWB today is not used as a fast data pipe. It is built for reliable time-of-flight ranging, secure distance measurement and consistent performance in indoor environments with reflections and people moving around.


#### 1.2 Typical Uses

UWB has moved from industrial positioning systems into everyday devices. It now appears in phones, access badges, tags and vehicles. Because it determines real physical distance rather than relying on received signal strength, it is much harder to spoof and more stable than BLE for ranging.

Common use cases include:

* Finding small tags and devices with high accuracy (Apple Find My)
* Secure proximity unlock for vehicles and smart locks
* Indoor positioning for robots, machinery and AGVs
* Worker safety and zone awareness in industrial facilities
* Precise relative positioning in AR and VR systems
