# Wireless Networking and Mobile Computing

### 1. Introduction

Peer-to-peer (P2P) and mesh networking enable direct communication between devices without relying on fixed infrastructure or central servers.
Each device can act as both sender and receiver, forming flexible, self-organizing networks that can operate even when mobile coverage or Internet access is unavailable.

These technologies are important in environments where connectivity is limited, temporary, or needs to remain private.
They also support local communication setups such as smart home systems, local file sharing, or office collaboration networks that do not depend on the Internet.

Mesh networking extends this concept by allowing devices to forward messages on behalf of others.
This means that even if two devices are not directly within range, data can still travel across multiple intermediate nodes.
Together, P2P and mesh systems create the foundation for independent, resilient, and privacy-preserving communication networks.

### 2. Peer-to-Peer Communication Protocols

Modern smartphones and computers provide several ways to establish direct device-to-device connections.
The main protocols used for this purpose are Wi-Fi Aware, Bluetooth Low Energy (BLE), and Bonjour.
Each serves a different purpose, optimized for range, energy use, and type of data transfer.



### 2.1 Wi-Fi Aware

Wi-Fi Aware, or Neighbor Awareness Networking (NAN), allows nearby devices to discover and communicate directly using Wi-Fi without an access point.
It was introduced on Android and is now also supported on iOS, enabling cross-platform discovery and data exchange.

#### Purpose

Wi-Fi Aware was built to enable local discovery and communication for file sharing, multiplayer games, smart home control, and context-aware apps.
Devices advertise services and connect automatically when in proximity, using the same radio used for regular Wi-Fi.

#### Strengths

* High throughput and long range, using standard Wi-Fi hardware.
* Efficient background discovery without heavy scanning.
* Well suited for mobile ad hoc or small mesh systems.
* Works on both Android and iOS, though APIs differ slightly.

#### Limitations

* Connection setup can be slow under load.
* Android limits simultaneous sockets (around five active sessions).
* Some devices still lack full NAN support or restrict background operation.
* No native routing or persistence after disconnection.

Wi-Fi Aware is an effective foundation for direct, high-bandwidth, short-range networking, especially when devices appear and disappear dynamically.



### 2.2 Bluetooth Low Energy (BLE)

Bluetooth Low Energy provides short-range, ultra-low-power communication, ideal for periodic or lightweight data exchange.
It operates in the 2.4 GHz ISM band and is designed for devices that need to advertise small pieces of information efficiently.

#### Purpose

Created for sensors, wearables, and IoT beacons, BLE focuses on energy efficiency and simplicity rather than throughput.

#### Strengths

* Extremely low energy consumption for long-term operation.
* Supported across almost all mobile and embedded platforms.
* Supports both connectionless broadcasting and stable GATT-based sessions.
* Rotating device identifiers improve privacy.

#### Limitations

* Very limited data rate and range compared to Wi-Fi.
* High latency and small packet sizes.
* Multi-hop or group routing is not supported.
* Simultaneous connections can be unreliable on some platforms.

BLE is best for advertising, discovery, and lightweight messaging, not continuous or bulk data transfer.



### 2.3 Bonjour + NWConnection

Bonjour is Apple’s implementation of Zero Configuration Networking (mDNS and DNS-SD), allowing devices to automatically discover services and establish local connections on the same network.
When combined with NWConnection and NWListener in Apple’s Network framework, it provides a secure, socket-like P2P transport layer.

#### Purpose

Bonjour makes it easy to find and connect devices in the same Wi-Fi network or direct Wi-Fi link.
It powers applications like AirPlay, AirDrop, and collaborative tools that need instant local communication.

#### Strengths

* Seamless service discovery and automatic connection setup.
* Secure data exchange using TLS by default.
* Handles multiple concurrent connections with minimal setup.
* Works consistently across macOS, iOS, and iPadOS.

#### Limitations

* Restricted to the local network segment.
* Multicast DNS may be blocked on some networks.
* No multi-hop routing or cross-platform compatibility.

Bonjour with NWConnection is reliable for local, encrypted communication between Apple devices, and serves as a solid transport base for higher-level mesh or DTN layers.

| Feature          | Wi-Fi Aware                  | BLE                        | Bonjour + NWConnection  |
| - | - | - | - |

### 3. Mesh Networking

Mesh networking extends peer-to-peer communication into a multi-hop structure. Instead of connecting only to nearby peers, each device also helps forward data for others. This allows messages to reach nodes that are outside direct radio range and improves overall resilience. If one node moves or fails, data can still flow through alternate paths.

Mesh networks can be infrastructure-free, meaning no routers, access points, or Internet links are required. Every participant contributes to connectivity, forming a self-healing, self-organizing network.

#### 3.1 How a Mesh Network Works

Each node maintains a small table of neighbors discovered through short beacon messages.
When a message is created, it can either be broadcast across all links or routed intelligently through selected paths.
As nodes join, leave, or move, the topology updates automatically.

Key properties:

* Redundancy: multiple alternative routes between any two devices.
* Scalability: new nodes expand coverage without reconfiguration.
* Adaptivity: route tables update as the environment changes.
* Decentralization: no single point of failure or control.

Mesh networking is especially valuable for emergency communication, rural connectivity, or temporary networks that must form instantly and operate without coordination.



#### 3.2 Routing Approaches

Several routing strategies are used depending on network size, mobility, and power constraints:

| Type                             | Description                                                                                            | Typical Use                                       |
| - | - | - |
| Flooding / Epidemic Routing  | Every node rebroadcasts received messages to all neighbors until a time limit or hop count expires.    | Small networks, delay-tolerant or emergency cases |
| Link-State Routing           | Each node maintains knowledge of all network links and computes shortest paths (e.g., Dijkstra).       | Stable topologies, medium-sized meshes            |
| Distance-Vector Routing      | Nodes exchange only their local distance metrics and choose next hops based on neighbors’ information. | Moderate mobility, low processing cost            |
| On-Demand / Reactive Routing | Routes are discovered only when data needs to be sent (e.g., AODV).                                    | Mobile ad hoc networks with limited resources     |
| Hybrid Routing               | Combines proactive local tables with on-demand global search.                                          | Large or variable-size meshes                     |

Each has a trade-off between latency, traffic overhead, and route stability.
In highly dynamic environments, periodic link updates can overwhelm the channel, so lightweight or probabilistic routing is often used.



#### 3.3 Building a Mobile Mesh System

A practical mesh implementation requires three key layers:

1. **Discovery and Link Layer**
   Handles neighbor detection and connection setup. This can be based on Wi-Fi Aware, Wi-Fi Direct, or BLE, depending on hardware capabilities.

2. **Routing Layer**
   Maintains paths between peers and decides how packets should travel. Link-state and distance-vector methods are common here.

3. **Transport and Application Layer**
   Provides reliable delivery, acknowledgments, and encryption. Applications can then send messages, files, or status updates transparently.

In mobile environments, it’s common to keep routing lightweight and stateless, focusing on short messages and limited-hop forwarding to reduce congestion and battery drain.

#### 3.4 Challenges and Design Trade-Offs

Mesh networks face specific limitations:

* Mobility and link churn: frequent reconfiguration can cause route instability.
* Power constraints: continuous discovery and transmission drain batteries quickly.
* Congestion: too many rebroadcasts can flood the network.
* Security: without a central authority, identity and encryption must be managed carefully.
