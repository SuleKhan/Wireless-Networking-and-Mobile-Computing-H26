# Wireless Networking and Mobile Computing

### 1. Zigbee

Zigbee builds on IEEE 802.15.4-2004 and targets long-lived, low-data networks where devices might run for months on AA batteries. It is designed to form and maintain large, self-organizing mesh networks that can scale to roughly 65 536 nodes and operate for years without human intervention. Data rates are modest at about 250 kb/s, spread across 27 channels in the 2.4 GHz and 868/915 MHz bands. The technology is optimized for tiny payloads, predictable sleep cycles and reliable multi-hop forwarding rather than throughput or quality-of-service. The result is low cost, low setup overhead and long-term autonomy in sensor and control deployments.

#### 1.1 Network Structure and Roles

Zigbee defines three device roles. The coordinator forms the network, assigns parameters and can bridge to other systems; only one exists within a Zigbee network. Routers forward traffic and extend coverage, creating the mesh backbone. End devices attach to a parent (either the coordinator or a router) and can sleep deeply because they are not responsible for forwarding. This structure supports large-scale trees and meshes: a thermostat talking to a heating controller, motion sensors feeding a security hub, or energy meters reporting to a gateway. Each node does a small job, and together they create a quiet, persistent nervous system for a building or industrial site with minimal human involvement.

#### 1.2 Radio and PHY

802.15.4 uses direct sequence spread spectrum and narrow channels to balance robustness with low power draw. In the 2.4 GHz band Zigbee operates with 16 channels, and regions like North America and Europe add sub-GHz channels around 868 and 915 MHz. Those lower bands trade data rate for improved penetration through walls and longer range.

Modulation options include BPSK, ASK and O-QPSK, with chip rates and symbol rates tailored to stay efficient at tiny payloads. The radio is designed to wake, send a short frame, and sleep again instead of running continuously. Transmission power is small, but mesh forwarding compensates by letting many tiny links form one large fabric.

#### 1.3 MAC Behavior and Power Use

Zigbee MAC behavior can adapt to the deployment. In non-beacon networks, nodes contend for the channel using unslotted CSMA-CA and rely on acknowledgements to confirm packets. It behaves simply and flexibly, suitable when devices wake unpredictably.

Beacon-enabled mode adds more structure. Coordinators send beacons defining time slots; devices can sleep between them and wake only at assigned moments. Guaranteed Time Slots exist for latency-sensitive traffic, and this scheduling helps battery-powered nodes last years.
