### 1. Near Field Communication (NFC)

NFC is a very short-range radio system that enables secure communication between devices only a few centimeters apart. It operates at 13.56 MHz with tight bandwidth and low power, which makes the range naturally limited and hard to intercept. Unlike most radios, proximity is part of the security model. You bring two devices together and the interaction happens right there, almost like tapping objects to wake them up digitally.

A major difference from Bluetooth and Wi-Fi is that NFC does not require pairing, scanning or manual setup. Instead, the radio link establishes almost instantly when two coils couple magnetically. One device can power the other over this field, so passive NFC tags do not need batteries. That makes NFC ideal for stickers, cards, key fobs, toys and packaging. Data rates range roughly from 106 to 424 kb/s, which is more than enough for identifiers, configuration payloads or small structured messages.


#### 1.1 Modes and Capabilities

NFC devices can act in different roles. In reader mode, a phone or terminal energizes and communicates with a passive tag that stores a small amount of data, usually under 10 kB. In card emulation mode, a phone behaves like a secure NFC card, which is how payment systems and access badges work. There is also a peer-to-peer mode, where two active devices exchange small packets when touched or tapped together.

The technology follows ISO/IEC 18000-3 and related standards in the NFC Forum stack. Security and identity functions can be anchored in secure elements or trusted execution environments, which gives NFC its trustworthiness in payment and ticketing systems. The limited range is not a drawback here. It provides physical intent: if you tap, you meant to interact.

#### 1.2 Common Uses

NFC has quietly become part of everyday interaction patterns. It appears in transit tickets, bank cards, passports, hotel keys and product authentication. Phones read tags for pairing accessories, launching apps, unlocking scooters or checking in at events. In logistics, NFC tracks high-value items and provides tamper evidence. In toys, NFC tags trigger in-game content or unlock characters without batteries or wires.

Examples you see every day:

* Contactless payments and transit cards
* Secure building access badges
* e-Passports and identity documents
* Pairing headphones or printers by tapping
* Smart packaging and anti-counterfeit labels
* Rental scooters and shared mobility check-in
* Game figurines and toy-to-digital experiences

The user experience is consistent: bring two things together, something happens, then both devices go silent again.

#### 1.3 Why NFC Instead of BLE?

BLE could technically transfer the same amount of data, but NFC wins when intent, proximity and trust matter. There is no scanning delay, minimal configuration, and the physical action is part of the protocol. NFC tags also cost pennies, do not need batteries and can be embedded in paper, plastic or fabric. That combination makes NFC the simplest way to give everyday objects a tiny touch-based digital identity.

In short, NFC is not built for range or speed. It is built for secure, deliberate contact-based interaction, and that is why it works in wallets, doors, passports and product labels rather than in streaming or real-time systems.
