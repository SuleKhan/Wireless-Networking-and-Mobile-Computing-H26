# Wireless Networking and Mobile Computing

### 1. Game Theory and Mobile Resource Sharing

Game theory provides a structured way to analyze how independent users interact when they share limited communication resources. In mobile and wireless systems, many users compete for bandwidth, power, or time slots. Each participant, or player, acts rationally to maximize its own performance based on what it knows about others. The resulting system behavior emerges from these local, self-interested decisions.

In mobile computing, this framework helps describe how fairness, cooperation, and competition evolve naturally in dynamic, decentralized environments. A network’s overall efficiency often depends less on centralized control and more on how individual users adapt their strategies over time.

#### 1.1 Spectrum Sharing Games

When several transmitters share the same spectrum, their repeated attempts to access the channel can be viewed as a repetitive game. Each user’s goal is to achieve acceptable throughput and delay, while minimizing interference and collisions. The outcome of one round influences strategies in later rounds.

In such evolutionary games, strategies that perform well tend to spread across the population. Over time, systems may converge to a stable pattern where no single user benefits by changing its strategy alone. This steady state corresponds to a Nash Equilibrium in the context of wireless access.

Measured QoS values, such as throughput and delay, represent the observed payoffs for each user after each stage game. Repetition allows adaptation and learning, turning random access into an evolving system of competition and cooperation.

#### 1.2 Game Theory Concepts in Mobile Computing

Game-theoretic reasoning helps analyze how self-interested devices interact in networks.
Key questions include:

* How does rational self-interest lead to predictable network behavior?
* Under what conditions do independent users cooperate instead of compete?
* What trade-offs exist between fairness and efficiency?

Important terms:

* **Utility function:** quantifies how much benefit or satisfaction a player gains from a given outcome.
* **Payoff:** the achieved utility value, often related to QoS such as throughput or latency.
* **Nash Equilibrium:** a state where no player can unilaterally improve its payoff.
* **Pareto efficiency:** a condition where no player can be made better off without making another worse off.

Cooperation in such systems does not necessarily arise from altruism but from long-term self-interest. Even non-cooperative games may produce cooperative outcomes if players recognize that repeated defection leads to mutual loss.


#### 1.3 Utility Functions


Each player, indexed by $i$, aims to satisfy its own performance requirements. The **utility function** $U_i$ describes how well those requirements are met. Players attempt to maximize their observed payoff when measured performance meets or exceeds their targets.

Let:

* demand $= f(t)$ be the time-varying desired performance
* requirement be a constant target value
* observation $= f(n)$ be measured throughput $\Theta$ and delay $\Delta$

A high utility indicates that the observed QoS values meet or surpass user expectations. The function may penalize delay or reward higher throughput depending on the application. A common example is

$$
U_i(\Theta, \Delta) = w_1 \log(\Theta_i) - w_2 \Delta_i
$$

where $w_1$ and $w_2$ are weights expressing how much the player values throughput versus delay.

Players adjust their strategies to maximize $U_i$, trading off between aggressive channel access (higher $\Theta$ but higher $\Delta$ for others) and cooperative sharing that stabilizes the overall system.

#### 1.4 Pareto Efficiency and the Bargaining Domain

When rational players repeatedly interact, they often converge toward a Nash Equilibrium (NE). At this point, no single player can gain by changing its own action alone.

If no other action profile exists with strictly higher payoffs for all players, then this equilibrium is Pareto efficient.
However, some equilibria are stable but not efficient; all players might improve together if they could coordinate or bargain. The space between the NE and the Pareto frontier defines the bargaining domain, where cooperative strategies can provide better overall results.

Repeated interaction and feedback allow rational agents to learn and sometimes evolve cooperative behavior even without explicit negotiation. Over time, this dynamic equilibrium may stabilize into predictable sharing patterns.

#### 1.5 Cooperation and Repeated Interaction

In mobile resource sharing, long-term cooperation can emerge from purely self-interested behavior. When players interact repeatedly and remember past outcomes, strategies evolve that reward cooperation and punish defection.

Such dynamics can lead to balanced performance even without external enforcement.
This forms the basis for non-cooperative cooperation: systems that self-stabilize through strategic reciprocity rather than regulation.

In practice, this mirrors mobile nodes adapting contention windows or power levels to maintain coexistence with minimal interference.


#### 1.6 Types of Social Actions

Sociologist Max Weber identified four types of motivations for human behavior, which also describe how autonomous agents might act in network games:

1. **Affective social action**: behavior driven by emotion or immediate reaction.
2. **Traditional social action**: following established norms or habits.
3. **Technocratic rational action** (*zweckrational*): goal-oriented behavior guided by logic and efficiency.
4. **Value-oriented rational action** (*wertrational*): decisions guided by values or principles, even at personal cost.

In network contexts, value-oriented actors (for example, cooperative nodes programmed for fairness) can reduce the need for external regulation or central control.


#### 1.7 Example: Tit-for-Tat in Repeated Games

A well-known example of cooperative strategy is **Tit-for-Tat** in the repeated Prisoner’s Dilemma.
The rule is simple: start by cooperating, then replicate the opponent’s previous move.

This strategy combines fairness with forgiveness. It encourages cooperation by rewarding good behavior and discourages defection through immediate reciprocity.

Applied to wireless systems, nodes following Tit-for-Tat might initially share spectrum generously, but will reduce cooperation if others behave selfishly, gradually reestablishing equilibrium. Such adaptive reciprocity can sustain balanced network operation even without centralized control.
