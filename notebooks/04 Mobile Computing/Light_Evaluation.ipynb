# Wireless Networking and Mobile Computing

### 1. Microcontroller Testbed

This evaluation uses simple off-the-shelf microcontrollers driving and sensing a single LED. The same LED provides illumination, transmits data by brief intensity changes, and (when reverse-biased) senses incoming light as a low-cost receiver.

  - ATmega328P (8-bit, 32 KB flash, 2 KB RAM, multiplexed 8-ch ADC)
  - ARM boards

### 2. VLC Evaluation

#### 2.1 Single-Link

A one-to-one line-of-sight link on a bench rail with adjustable distance. 

<center>
  <img src="../../Admin/light/vlc_single.png" width="40%">
</center>

For distances up to 2 meters, the throughput is constant for all packet sizes. Beyond this range, throughput collapes because of increased error probability due to bad reception. The peak measured throughput is 800 bit/s with 100-byte payloads.

<center>
  <img src="../../Admin/light/single_throughput.png" width="40%">
  <img src="../../Admin/light/single_delays.png" width="41%">
</center>

#### 2.2 Multi-Node

Star-style arrangement with all nodes mutually visible to avoid hidden-station effects. There is a smaller packet size due to the increased error probability. 

<center>
  <img src="../../Admin/light/vlc_network.jpg" width="40%">
</center>

Up to 5 transmitters have total throughput remain stable, as CSMA/CA and idle resync handle contention well. With more devices, retries and missed SFDs increase, so the throughput drops. The measured network throughput is up to 500 bit/s.
