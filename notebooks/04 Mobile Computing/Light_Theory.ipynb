# Wireless Networking and Mobile Computing

### 1. Visible Light Communication (VLC)

Visible Light Communication uses variations in LED light intensity to transmit data. The changes happen faster than human eyes can detect, so lighting and communication occur at the same time. VLC is being explored for next-generation indoor networking, short-range device communication, and underwater links where radio signals struggle.

#### 1.1 Historical Starting Point: The Photophone

The first documented VLC device was the Photophone, invented by Alexander Graham Bell in 1880. It transmitted spoken audio ~200 meters by modulating sunlight onto a mirror and directing it to a receiver. Although impractical at the time, it demonstrated the principle of sending information using visible light. Modern LEDs and microelectronics revived that concept more than 100 years later.

#### 1.2 Why Use Visible Light?

The visible spectrum is extremely wide (hundreds of THz) and unlicensed. With LED lighting deployed everywhere, VLC can piggyback on illumination systems. Light is also well-suited for environments where radio is restricted or ineffective.

Reasons VLC is attractive:

* huge unlicensed bandwidth
* harmless and already integrated into consumer spaces
* does not interfere with RF systems
* strong underwater performance
* possible support in future 6G and Li-Fi systems

Early research has also suggested VLC inside vehicles, airplanes, hospitals, wearable devices and toys — places where RF is noisy, regulated or power-constrained.

#### 1.3 LED Principles for Communication

LEDs emit light when forward-biased, and they can also detect light when reverse-biased, similar to photodiodes (though less sensitive). That dual nature simplifies small VLC circuits. In basic systems, an LED can serve as both transmitter and receiver, modulating its brightness to send data and sensing incoming light to receive data.

The key idea is intensity modulation: the light does not change color, only brightness at very high speed. The human eye perceives a steady beam while devices see rapid digital pulses. LEDs are useful here because they are not only emitters. When reverse-biased, an LED behaves like a simple light sensor. Incoming photons free charge carriers in the LED junction, and the stored charge leaks away faster when light hits it. A microcontroller can monitor that voltage change, so the same component that sends light can also detect it, just with lower sensitivity than a dedicated photodiode.
