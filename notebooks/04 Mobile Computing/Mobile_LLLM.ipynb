# Wireless Networking and Mobile Computing

### 1. Local Large Language Models

#### 1.1 From Cloud to Local LLMs

Traditional LLMs usually rely on powerful cloud infrastructure to run efficiently at scale. Local LLMs make it possible to run models directly on personal devices or edge hardware instead. This shift enables greater control over data and computation, and it also supports offline operation with less dependency on external services.

#### 1.2 Local vs Cloud Based LLMs

Local LLMs give full control over privacy and security because data can stay on the device. They also work without internet access and can reduce operational costs for small scale or personal use. Cloud based LLMs provide access to very large and advanced models with high computational performance and scalability, and they benefit from continuous updates and fine tuning by providers. The trade off is that local use favors autonomy and privacy but is limited by device compute, while cloud use favors performance and convenience but reduces control over data and infrastructure.

#### 1.3 What Are Large Language Models?

Large language models are AI systems that process and generate text by transforming an input sequence such as a question, instruction, or statement into an output sequence such as a continuation, answer, or reformulation. They operate token by token by predicting the next most likely word. Over time, they evolved from narrow task systems like automated translation into general purpose models that learn broad text patterns from massive datasets. Common applications include translation, code generation, creative writing, question answering, and summarization.

#### 1.4 What Are Embeddings?

Embeddings are numerical representations of language that map words, sentences, or documents into a continuous vector space. In this space, items with similar meaning tend to appear close together. The distance and direction between vectors can encode similarity such as near synonyms, and also encode relationships and analogies that capture higher level semantic structure.
