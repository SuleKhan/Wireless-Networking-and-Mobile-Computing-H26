### 1. Stochastic Distributions
#### 1.1 Experiment Setup

We generate a number of lists of a certain size (number of samples). The list contains independent and identically distributed (i.i.d.) random numbers.
 - **Independent**: Each random variable (or sample) is independent of the others, meaning the outcome of one does not affect the others.
 - **Identically Distributed**: All random variables follow the same probability distribution.

```python
# --- Data Generation: create random data based on student legi number ---
low = (10 * legi_nr) / 10000
high = (11 * legi_nr) / 10000

random_data = {}
for i in range(n):
    random_data[f"data_random{i+1:02}"] = np.random.uniform(low, high, size).flatten()
```

#### 1.2 Plot Time Series

A time series plot visualizes how the values of each random data list change across their sample indices, revealing trends (or randomness) over time.

```python
# --- Plotting: visualize the generated random data ---
fig, ax_time = plt.subplots(figsize=(8, 5))
theLineWidth = 0.5

for i, (key, value) in enumerate(random_data.items()):
    ax_time.plot(value, label=f'Random Data List {i+1:02}', alpha=1, linewidth=theLineWidth)

ax_time.set_xlabel('Index')
ax_time.set_ylabel('Random Values')
ax_time.set_title('Time Series')
ax_time.grid(True)
ax_time.set_facecolor((.2, 0.2, 0.3))
ax_time.legend(fancybox=True, framealpha=1, loc="upper left", facecolor=(.6, 0.61, 0.8))
ax_time.autoscale(enable=True, axis='both')

plt.tight_layout()
```

#### 1.3 Plot Histograms

A histogram is a graphical representation that shows the distribution of a dataset by grouping values into bins and displaying the frequency of values in each bin.

```python
# --- Plotting: histogram of the generated random data ---
fig, ax_hist = plt.subplots(figsize=(8, 5))
theLineWidth = 2
theNumberOfBins = 100

for i, (key, value) in enumerate(random_data.items()):
    ax_hist.hist(value, bins=theNumberOfBins, label=f'Random Data List {i+1:02}', alpha=0.7)

ax_hist.set_xlabel('Value')
ax_hist.set_ylabel('Frequency')
ax_hist.set_title('Histogram')
ax_hist.grid(True)
ax_hist.set_facecolor((.2, 0.2, 0.3))
ax_hist.legend(fancybox=True, framealpha=1, loc="upper left", facecolor=(.6, 0.61, 0.8))
ax_hist.autoscale(enable=True, axis='both')

plt.tight_layout()
```

#### 1.4. Densities
We explain and visualize the probability distribution (PDF) and complementary cumulative distribution (CCDF) for a random data list. The latter is often used in wireless communication to show the probability of a random variable being greater than a particular value (e.g the probability $P[X > x]$ of a packet observing a certain delay $x$). Therefore it starts with $1$, representing a $100$% probability that the delay is greater or equal to $0$.



```python
# --- Plotting: PDF and Complementary CDF of one of the generated random data lists ---
fig, (ax_pdf, ax_cdf) = plt.subplots(1, 2, figsize=(14, 5))
theLineWidth = 2

# Choose one list to draw the PDF and CDF (for example, the first list)
selected_list = random_data["data_random01"]

# PDF (left)
kde = gaussian_kde(selected_list)
x_pdf = np.linspace(low, high, 300)
ax_pdf.plot(x_pdf, kde(x_pdf), color='w', linewidth=theLineWidth, label='PDF')
ax_pdf.set_xlabel('Value')
ax_pdf.set_ylabel('Density')
ax_pdf.set_title('PDF')
ax_pdf.grid(True)
ax_pdf.set_facecolor((.2, 0.2, 0.3))
ax_pdf.legend(fancybox=True, framealpha=1, loc="upper right", facecolor=(.6, 0.61, 0.8))
ax_pdf.autoscale(enable=True, axis='both')

# Complementary CDF (right)
sorted_data = np.sort(selected_list)
cdf = np.arange(1, size + 1) / size
ccdf = 1 - cdf
ax_cdf.plot(sorted_data, ccdf, color='g', linewidth=theLineWidth, label='Complementary CDF')
ax_cdf.set_xlabel('Value')
ax_cdf.set_ylabel('Complementary Probability')
ax_cdf.set_title('Complementary CDF')
ax_cdf.grid(True)
ax_cdf.set_facecolor((.2, 0.2, 0.3))
ax_cdf.legend(fancybox=True, framealpha=1, loc="upper right", facecolor=(.6, 0.61, 0.8))
ax_cdf.autoscale(enable=True, axis='both')

plt.tight_layout()
```

#### 1.5 Central Limit Theorem
> When you take the sum (or average) of a large number of independent and identically distributed (i.i.d.) random variables, regardless of the original distribution of these variables, their normalized sum tends toward a normal (Gaussian) distribution as the number of variables increases.

$\implies$ The sum or average of a large number of i.i.d random variables tends to follow a normal distribution.

This is an important concept in probability theory and statistics, and highly relevant in practical communication scenarios with a large number of noise sources, interferers, or other independent processes.


```python
# --- Plotting: PDF and Complementary CDF of the element-wise sum of all generated random data lists ---
sum_data = np.zeros_like(random_data["data_random01"])
for i in range(1, n+1):
    sum_data += random_data[f"data_random{i:02}"]

# PDF and CCDF for the summed data
fig, (ax_pdf, ax_cdf) = plt.subplots(1, 2, figsize=(14, 5))
theLineWidth = 2

# PDF (left)
kde_sum = gaussian_kde(sum_data)
x_sum = np.linspace(np.min(sum_data), np.max(sum_data), 300)
ax_pdf.plot(x_sum, kde_sum(x_sum), color='w', linewidth=theLineWidth, label='PDF (Sum)')
ax_pdf.set_xlabel('Value')
ax_pdf.set_ylabel('Density')
ax_pdf.set_title('PDF (Element-wise Sum)')
ax_pdf.grid(True)
ax_pdf.set_facecolor((.2, 0.2, 0.3))
ax_pdf.legend(fancybox=True, framealpha=1, loc="upper right", facecolor=(.6, 0.61, 0.8))
ax_pdf.autoscale(enable=True, axis='both')

# Complementary CDF (right)
sorted_sum = np.sort(sum_data)
cdf_sum = np.arange(1, len(sum_data) + 1) / len(sum_data)
ccdf_sum = 1 - cdf_sum
ax_cdf.plot(sorted_sum, ccdf_sum, color='g', linewidth=theLineWidth, label='Complementary CDF (Sum)')
ax_cdf.set_xlabel('Value')
ax_cdf.set_ylabel('Complementary Probability')
ax_cdf.set_title('Complementary CDF (Element-wise Sum)')
ax_cdf.grid(True)
ax_cdf.set_facecolor((.2, 0.2, 0.3))
ax_cdf.legend(fancybox=True, framealpha=1, loc="upper right", facecolor=(.6, 0.61, 0.8))
ax_cdf.autoscale(enable=True, axis='both')

plt.tight_layout()
```
