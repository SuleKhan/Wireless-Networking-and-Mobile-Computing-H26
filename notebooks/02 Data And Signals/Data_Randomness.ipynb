#### 1.1 Randomness and PRNGs

A pseudo-random number generator (PRNG) is a deterministic function $x_n = f(s, n)$ that maps a seed $ s $ and an index $ n $ to an output value. The generator maintains an internal state that evolves deterministically:

$$
s_{n+1} = g(s_n), \quad x_n = h(s_n).
$$

Because $g$ and $h$ are deterministic, using the same seed always produces the same sequence:

$$
\forall n: x_n(s) = x_n(s),
$$

while different seeds usually lead to distinct sequences:

$$
\forall n: x_n(s_1) \neq x_n(s_2) \quad \text{for } s_1 \neq s_2.
$$

In practice, reproducibility follows from this property. The quality of a PRNG depends on its period length, uniformity, and independence, which together determine how well its output approximates true randomness.


```python
# --- Generate & Save: create data arrays and write separate CSVs ---
csv_random_path = Path(csv_random_path)
csv_constant_path = Path(csv_constant_path)

seed = legi_nr

rng = np.random.default_rng(seed)
size = int(size)

# arrays
data_ones01 = np.ones(size) * 0.5
data_random01 = rng.random(size)

# Save random data to its CSV
df_random = pd.DataFrame({'data_random01': data_random01})
df_random.to_csv(csv_random_path, index=False)

# Save constant data to its own CSV
df_const = pd.DataFrame({'data_ones01': data_ones01})
df_const.to_csv(csv_constant_path, index=False)

print(f'Saved random data to {csv_random_path.resolve()}')
print(f'Saved constant data to {csv_constant_path.resolve()} (seed={seed}, size={size})')
```

#### 1.3 Seed Values
A fixed seed yields identical runs, while seeding from entropy sources introduces non-deterministic variation.

```python
# --- Demo: show which sequences are identical when using the same seed ---
seed_base = seed if seed is not None else 1
seed_other = None

# Create two independent RNG instances with the SAME seed for pair A
rngA1 = np.random.default_rng(seed_base)
rngA2 = np.random.default_rng(seed_base)

# Create two independent RNG instances with seed_other (None -> non-deterministic)
rngB1 = np.random.default_rng(seed_other)
rngB2 = np.random.default_rng(seed_other)

# Generate sequences
x1 = rngA1.random(size)
x2 = rngA2.random(size)
x3 = rngB1.random(size)
x4 = rngB2.random(size)

# Short previews (5 entries) rounded to 3 decimals
s1 = np.round(x1[:5], 3).tolist()
s2 = np.round(x2[:5], 3).tolist()
s3 = np.round(x3[:5], 3).tolist()
s4 = np.round(x4[:5], 3).tolist()

print(f'(seed_base = {seed_base}, seed_other = {seed_other})')

print()
print(f'x1: {s1} ...')
print(f'x2: {s2} ...')
print(f'x3: {s3} ...')
print(f'x4: {s4} ...')
```
