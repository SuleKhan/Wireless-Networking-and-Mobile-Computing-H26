#### 2.1.1 Signal Generation

```python
# --- Generating Analog Signals ---
sampling_rate = 10 * max(freq_a_Hertz, freq_b_Hertz)
duration = 20
t_min_sec = 0
t_max_sec = t_min_sec + duration
number_of_samples = sampling_rate * (t_max_sec-t_min_sec)

sampling_interval_sec = (t_max_sec-t_min_sec)/number_of_samples

# Time vector
time_sec = np.linspace(0, duration, int(
    sampling_rate * duration), endpoint=False)

# Harmonic signal
signal_a = np.sin(2 * np.pi * freq_a_Hertz * time_sec)
signal_b = np.sin(2 * np.pi * freq_b_Hertz * time_sec)

# add leading and trailing zeros to the data
time_sec = np.concatenate((np.linspace(t_min_sec - (t_max_sec-t_min_sec), t_min_sec, round(number_of_samples)), time_sec, np.linspace(t_max_sec, t_max_sec+(t_max_sec-t_min_sec), round(number_of_samples))))
signal_a = np.concatenate((np.array(np.zeros(len(signal_a))), signal_a, np.array(np.zeros(len(signal_a)))))
signal_b = np.concatenate((np.array(np.zeros(len(signal_b))), signal_b, np.array(np.zeros(len(signal_b)))))
```

#### 2.1.3 Analog Signals in Frequency Domain
A Fourier analysis is a method for expressing a function as a sum of periodic components. When  a signal and its Fourier transform are replaced with discretized counterparts, it is called the discrete Fourier transform (DFT).

A fast algorithm for computing the DFT is referred to as the ***Fast Fourier Transform (FFT)***. Because the discrete Fourier transform separates its input into components that contribute at discrete frequencies, it has applications in digital signal processing. The discretized ***input*** to the transform is referred to as a ***signal***, which exists in the ***time domain***. The ***output*** is referred to as the ***spectrum*** (or transform) and exists in the ***frequency domain***.

```python
# --- Signals in Frequency Domain ---
fft_a = (np.fft.fft(signal_a))
fft_b = (np.fft.fft(signal_b))

# Plotting
fig, ax_freq = plt.subplots(figsize=(8, 5))
theLineWidth = 2

ax_freq.cla()
ax_freq.grid(color="#AAAAAA", linewidth=.5, linestyle='-', alpha=1)
ax_freq.set_ylabel('signals in frequency domain ( |S(f)| )')
ax_freq.set_xlabel('frequency (Hz)')
ax_freq.axis([None, None, None, None])

freq_Hz = np.fft.fftfreq(len(fft_a), sampling_interval_sec)
handle_Freq_a = ax_freq.plot(freq_Hz, abs(fft_a)/number_of_samples*2, color='m', linewidth=theLineWidth, label=[
                             'signal a: ' + str(freq_a_Hertz) + ' Hz'])
handle_Freq_b = ax_freq.plot(freq_Hz, abs(fft_b)/number_of_samples*2, color='y', linewidth=theLineWidth, label=[
                             'signal b: ' + str(freq_b_Hertz) + ' Hz'])
ax_freq.set_xlim(0, 1.5 * np.max([freq_a_Hertz, freq_b_Hertz]))
ax_freq.set_facecolor((.2, 0.2, 0.3))
plt.legend(fancybox=True, framealpha=1,
           loc="upper left", facecolor=(.6, 0.61, 0.8))
plt.tight_layout()
```

### 2.2 Digital Signals

#### 2.2.1 Digital Modulation

Digital modulation refers to the process of encoding digital information onto an analog carrier signal, such as a sine wave, for transmission over a communication medium. Unlike analog modulation, where the modulating signal is continuous, in digital modulation, the information is represented in discrete values (bits: 0s and 1s). This is commonly used in modern communication systems, such as Wi-Fi. Common types of digital modulation shift the signal amplitude with each symbole, or the signal phase, or a combination of phase and amplitude.

**Amplitude Shift Keying (ASK)**

>In ASK, the amplitude of the carrier wave is varied to represent the digital data. Mathematical representation:
>
 >- $s(t) = A_0 \cdot \cos(2 \pi f_c t) \quad \text{for 0}$
 >- $s(t) = A_1 \cdot \cos(2 \pi f_c t) \quad \text{for 1}$
>
>Where $ f_c $ is the carrier frequency, and $ A_0 $ / $ A_1 $ are the amplitudes representing binary 0 and 1, respectively.




```python
# --- Generating ASK Modulated Signals ---
carrier_frequency_a = freq_a_Hertz
carrier_frequency_b = freq_b_Hertz

t = np.linspace(0, 1/bit_rate, int(sampling_rate/bit_rate), endpoint=False)

# Generate amplitude modulated signal
ampl_signal_a = np.array([])
ampl_signal_b = np.array([])
data_signal = np.array([])

for bit in data_short:
    carrier_a = np.cos(2 * np.pi * carrier_frequency_a * t)
    carrier_b = np.cos(2 * np.pi * carrier_frequency_b * t)
    if bit == 0:
        modulated_bit_a = carrier_a * 0.3  # amplitude shift for 0
        modulated_bit_b = carrier_b * 0.3  # amplitude shift for 0
        data_signal_one_bit = np.ones(len(t))
    else:
        modulated_bit_a = carrier_a  # No amplitude shift for 1
        modulated_bit_b = carrier_b  # No amplitude shift for 1
        data_signal_one_bit = - np.ones(len(t))
    ampl_signal_a = np.concatenate((ampl_signal_a, modulated_bit_a))
    ampl_signal_b = np.concatenate((ampl_signal_b, modulated_bit_b))
    data_signal = np.concatenate((data_signal, data_signal_one_bit))
```

```python
# --- Plotting ASK Modulated Signals ---
fig, ax_time = plt.subplots(figsize=(8, 5))
theLineWidth = 1.5
ax_time.cla()
ax_time.grid(color="#AAAAAA", linewidth=.5, linestyle='-', alpha=1)
ax_time.set_ylabel('modulated signal in time domain, amplitude')
ax_time.set_xlabel('time (s)')
ax_time.axis([None, None, None, None])

ax_time.plot(np.linspace(0, len(ampl_signal_a)*sampling_interval_sec, len(ampl_signal_a)), ampl_signal_a, label=[
    'Amplitude-modulated signal a (' + str(freq_a_Hertz) + ' Hz)'], color='m', linewidth=theLineWidth)
ax_time.plot(np.linspace(0, len(data_signal)*sampling_interval_sec, len(data_signal)), .08*data_signal-1.1, label=[
    'data (' + str(bit_rate) + ' bit per second)'], color='g', linewidth=theLineWidth)

ax_time.set_facecolor((.2, 0.2, 0.3))
plt.legend(fancybox=True, framealpha=1,
           loc="upper left", facecolor=(.6, 0.61, 0.8))
plt.tight_layout()
plt.grid(True)
plt.show()

fig, ax_time = plt.subplots(figsize=(8, 5))
theLineWidth = 1.5
ax_time.cla()
ax_time.grid(color="#AAAAAA", linewidth=.5, linestyle='-', alpha=1)
ax_time.set_ylabel('modulated signal in time domain, amplitude')
ax_time.set_xlabel('time (s)')
ax_time.axis([None, None, None, None])

ax_time.plot(np.linspace(0, len(ampl_signal_b)*sampling_interval_sec, len(ampl_signal_b)), ampl_signal_b, label=[
    'Amplitude-modulated signal b (' + str(freq_b_Hertz) + ' Hz)'], color='y', linewidth=theLineWidth)
ax_time.plot(np.linspace(0, len(data_signal)*sampling_interval_sec, len(data_signal)), .08*data_signal-1.1, label=[
    'data (' + str(bit_rate) + ' bit per second)'], color='g', linewidth=theLineWidth)

ax_time.set_facecolor((.2, 0.2, 0.3))
plt.legend(fancybox=True, framealpha=1,
           loc="upper left", facecolor=(.6, 0.61, 0.8))
plt.tight_layout()
plt.grid(True)
plt.show()
```

**Phase Shift Keying (PSK)**

>In PSK, the phase of the carrier signal is shifted to represent binary data. Mathematical representation of Binary PSK (BPSK):
  >
 >- $s(t) = \cos(2 \pi f_c t) \quad \text{for 0}$
 >- $s(t) = \cos(2 \pi f_c t + \pi) \quad \text{for 1}$
  >
>Where $f_c $ is the carrier frequency. The phase is shifted by 0° for binary 0 and by 180° for binary 1.

```python
# --- Generating BPSK Modulated Signals ---
t = np.linspace(0, 1/bit_rate, int(sampling_rate/bit_rate), endpoint=False)

# Generate BPSK modulated signal
bpsk_signal_a = np.array([])
bpsk_signal_b = np.array([])
data_signal = np.array([])

for bit in data_short:
    carrier_a = np.cos(2 * np.pi * carrier_frequency_a * t)
    carrier_b = np.cos(2 * np.pi * carrier_frequency_b * t)
    if bit == 0:
        modulated_bit_a = -carrier_a  # 180-degree phase shift for 0
        modulated_bit_b = -carrier_b  # 180-degree phase shift for 0
        data_signal_one_bit = np.ones(len(t))
    else:
        modulated_bit_a = carrier_a  # No phase shift for 1
        modulated_bit_b = carrier_b  # No phase shift for 1
        data_signal_one_bit = - np.ones(len(t))
    bpsk_signal_a = np.concatenate((bpsk_signal_a, modulated_bit_a))
    bpsk_signal_b = np.concatenate((bpsk_signal_b, modulated_bit_b))
    data_signal = np.concatenate((data_signal, data_signal_one_bit))
```

```python
# --- Plotting BPSK Modulated Signals ---
fig, ax_time = plt.subplots(figsize=(8, 5))
theLineWidth = 1.5
ax_time.cla()
ax_time.grid(color="#AAAAAA", linewidth=.5, linestyle='-', alpha=1)
ax_time.set_ylabel('modulated signal in time domain, amplitude')
ax_time.set_xlabel('time (s)')
ax_time.axis([None, None, None, None])

ax_time.plot(np.linspace(0, len(bpsk_signal_a)*sampling_interval_sec, len(bpsk_signal_a)), bpsk_signal_a, label=[
    'BPSK-modulated signal a (' + str(freq_a_Hertz) + ' Hz)'], color='m', linewidth=theLineWidth)
ax_time.plot(np.linspace(0, len(data_signal)*sampling_interval_sec, len(data_signal)), .08*data_signal-1.1, label=[
    'data (' + str(bit_rate) + ' bit per second)'], color='g', linewidth=theLineWidth)

ax_time.set_facecolor((.2, 0.2, 0.3))
plt.legend(fancybox=True, framealpha=1,
           loc="upper left", facecolor=(.6, 0.61, 0.8))
plt.tight_layout()
plt.grid(True)
plt.show()

fig, ax_time = plt.subplots(figsize=(8, 5))
theLineWidth = 1.5
ax_time.cla()
ax_time.grid(color="#AAAAAA", linewidth=.5, linestyle='-', alpha=1)
ax_time.set_ylabel('modulated signal in time domain, amplitude')
ax_time.set_xlabel('time (s)')
ax_time.axis([None, None, None, None])

ax_time.plot(np.linspace(0, len(bpsk_signal_b)*sampling_interval_sec, len(bpsk_signal_b)), bpsk_signal_b, label=[
    'BPSK-modulated signal b (' + str(freq_b_Hertz) + ' Hz)'], color='y', linewidth=theLineWidth)
ax_time.plot(np.linspace(0, len(data_signal)*sampling_interval_sec, len(data_signal)), .08*data_signal-1.1, label=[
    'data (' + str(bit_rate) + ' bit per second)'], color='g', linewidth=theLineWidth)

ax_time.set_facecolor((.2, 0.2, 0.3))
plt.legend(fancybox=True, framealpha=1,
           loc="upper left", facecolor=(.6, 0.61, 0.8))
plt.tight_layout()
plt.grid(True)
plt.show()
```

**Quadrature Amplitude Modulation (QAM)**

> QAM combines both amplitude modulation and phase modulation. This allows for the transmission of multiple bits per symbol, enhancing data rates. 
>
> For example, **16-QAM** transmits 4 bits per symbol using 16 distinct combinations of phase and amplitude. 
> - Such a combination is referred to as a symbol. 
> - Here, 16 is the **Modulation Order (M)** (which in general determines the number of distinct symbols that can be transmitted)
> - The **Bit Rate** is the number of bits transmitted per second. 
> - The **Symbol Rate** is the number of symbols transmitted per second. This can differ from the bit rate based on the modulation order.

```python
# --- Generating QAM Modulated Signals ---
t = np.linspace(0, 1/bit_rate, int(sampling_rate/bit_rate), endpoint=False)

# Generate QAM (QPSK / 4-QAM) modulated signal
qam_signal_a = np.array([])
qam_signal_b = np.array([])
data_signal = np.array([])

# Group bits into pairs
for i in range(0, data_short_length, 2):
    bits = data_short[i:i+2]
    if len(bits) < 2:
        bits = np.append(bits, 0)  # pad if odd number of bits

    # Gray mapping for QPSK
    if np.array_equal(bits, [0,0]):
        I, Q = -1, -1
    elif np.array_equal(bits, [0,1]):
        I, Q = -1, +1
    elif np.array_equal(bits, [1,1]):
        I, Q = +1, +1
    elif np.array_equal(bits, [1,0]):
        I, Q = +1, -1

    carrier_a_I = np.cos(2 * np.pi * carrier_frequency_a * t)
    carrier_a_Q = np.sin(2 * np.pi * carrier_frequency_a * t)  # quadrature
    carrier_b_I = np.cos(2 * np.pi * carrier_frequency_b * t)
    carrier_b_Q = np.sin(2 * np.pi * carrier_frequency_b * t)

    modulated_bit_a = I * carrier_a_I + Q * carrier_a_Q
    modulated_bit_b = I * carrier_b_I + Q * carrier_b_Q

    qam_signal_a = np.concatenate((qam_signal_a, modulated_bit_a))
    qam_signal_b = np.concatenate((qam_signal_b, modulated_bit_b))
    data_signal = np.concatenate((data_signal, I*np.ones(len(t))))  # just show I-channel as reference
```

```python
# --- Plotting QAM Modulated Signals ---
fig, ax_time = plt.subplots(figsize=(8, 5))
theLineWidth = 1.5
ax_time.cla()
ax_time.grid(color="#AAAAAA", linewidth=.5, linestyle='-', alpha=1)
ax_time.set_ylabel('modulated signal in time domain, amplitude')
ax_time.set_xlabel('time (s)')
ax_time.axis([None, None, None, None])

ax_time.plot(np.linspace(0, len(qam_signal_a)*sampling_interval_sec, len(qam_signal_a)),
             qam_signal_a, label=['QAM-modulated signal a (' + str(freq_a_Hertz) + ' Hz)'],
             color='m', linewidth=theLineWidth)
ax_time.plot(np.linspace(0, len(data_signal)*sampling_interval_sec, len(data_signal)),
             .08*data_signal-1.1, label=['data (' + str(bit_rate) + ' bps)'],
             color='g', linewidth=theLineWidth)

ax_time.set_facecolor((.2, 0.2, 0.3))
plt.legend(fancybox=True, framealpha=1, loc="upper left", facecolor=(.6, 0.61, 0.8))
plt.tight_layout()
plt.grid(True)
plt.show()

# Plot QAM signal b
fig, ax_time = plt.subplots(figsize=(8, 5))
theLineWidth = 1.5
ax_time.cla()
ax_time.grid(color="#AAAAAA", linewidth=.5, linestyle='-', alpha=1)
ax_time.set_ylabel('modulated signal in time domain, amplitude')
ax_time.set_xlabel('time (s)')
ax_time.axis([None, None, None, None])

ax_time.plot(np.linspace(0, len(qam_signal_b)*sampling_interval_sec, len(qam_signal_b)),
             qam_signal_b, label=['QAM-modulated signal b (' + str(freq_b_Hertz) + ' Hz)'],
             color='y', linewidth=theLineWidth)
ax_time.plot(np.linspace(0, len(data_signal)*sampling_interval_sec, len(data_signal)),
             .08*data_signal-1.1, label=['data (' + str(bit_rate) + ' bps)'],
             color='g', linewidth=theLineWidth)

ax_time.set_facecolor((.2, 0.2, 0.3))
plt.legend(fancybox=True, framealpha=1, loc="upper left", facecolor=(.6, 0.61, 0.8))
plt.tight_layout()
plt.grid(True)
plt.show()
```
