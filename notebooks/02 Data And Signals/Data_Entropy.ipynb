### 1. Data

Data refers to facts, figures, or symbols that represent information. It can come in many forms, such as numbers, text, images, audio, video, or binary values. Data in its raw form usually has no meaning by itself until it is processed or interpreted. Once we understand the context, data becomes meaningful information.

The main difference between **analog** and **digital data** lies in how the information is represented and processed: 
- Analog data is represented by continuous signals. It can take any value within a range. For example, the sound of a person’s voice or temperature over time is an analog signal because it can vary smoothly and take any value within the limits of the system. The data is not limited to discrete steps, allowing for infinite possible values. For example, the voltage in a microphone signal might fluctuate continuously. Examples are sound waves, light intensity, temperature changes, and electrical signals.

- Digital data is represented by discrete signals, which means it can only take specific, limited values, often represented as binary values (0s and 1s). Digital data is quantized into a series of bits, and it cannot represent the infinite range of values that analog can. For example, in a digital audio file, the continuous sound wave is sampled at discrete intervals. Examples are digital images, computer files, and binary-encoded data.

### 2 Entropy

We now analyze the entropy of the three data sets above.

#### 2.1 Introduction

The entropy value reflects the randomness or unpredictability of the data values. It quantifies how much information is contained in the array, based on the distribution of the values. 

Low entropy means less uncertainty, more predictability. High entropy means more uncertainty, more diversity in the data. 

If data contains mostly the same values or very few different values, the entropy is low. This is because the data is predictable and contains less information. If the data values are evenly distributed, the entropy is high. This is because the data is less predictable and contains more information.

#### 2.2 Definition

The entropy $ H(X) $ of a random variable $ X $ with probabilities $ P(x) $ for each possible value $ x $ is calculated using Shannon's formula:

$$H(X) = - \sum P(x) \log_2 P(x), $$

where $ P(x) $ is the probability of a particular value $ x $ occurring in the array (i.e., the frequency of that value divided by the total number of elements). The base 2 logarithm is used to measure entropy in bits.

```python
# --- Entropy Calculation ---
def calculate_entropy(matrix_2d):
    values, counts = np.unique(matrix_2d.flatten(), return_counts=True)
    probabilities = counts / counts.sum()
    return entropy(probabilities, base=2)

print (f"Entropy of all ones {m}x{n} matrix:")
print(f"  - Entropy ignoring order: {round(calculate_entropy(data2d_ones),4)}")
print (f"Entropy of random {m}x{n} matrix:")
print(f"  - Entropy ignoring order: {round(calculate_entropy(data2d_rand),4)}")
print (f"Entropy of pattern {m}x{n} matrix:")
print(f"  - Entropy ignoring order: {round(calculate_entropy(data2d),4)}")
```

The code demonstrates that the classical entropy does not capture the information embedded into the order of the bits. The order does not make any difference: The second and the third data matrices result in the same entropy.



#### 2.2 Shannon Entropy over Bit Sequences
To calculate entropy considering the order of the bits, one approach is to treat a flattened one-dimensional array as a sequence and calculate the so-called Shannon entropy over bit sequences, which captures not just the randomness of individual bits but also the structure in how the bits are arranged in sequences.

The Shannon entropy calculates entropy based on the probability of different subsequences of a given length $ k $. 

A higher Shannon entropy implies more randomness or complexity in the sequences, while lower entropy implies more predictability. Given a sequence of bits, let $ X $ be a random variable representing a bit sequence of length $ k $ (called $ k $-mers). The Shannon Entropy $ H(X) $ of the bit sequences is defined as:

$$
H(X) = - \sum_{x \in X} P(x) \log_2 P(x),
$$

where
- $ X $ is the set of all possible bit sequences (or substrings) of length $ k $ found in the overall sequence
- $ P(x) $ is the probability of occurrence of each bit sequence $ x $ (i.e., the frequency of that bit sequence in the data, divided by the total number of sequences). 

As before, the logarithm is base 2 because entropy is measured in bits.

```python
# --- Sequence-based Entropy Calculation ---
def calculate_sequence_entropy(matrix_2d, k):
    flattened_array = matrix_2d.flatten()
    sequence = ''.join(map(str, flattened_array))
    # Count the occurrences of each k-length sequence (sliding window of size k)
    subsequences = [sequence[i:i + k] for i in range(len(sequence) - k + 1)]
    unique_subsequences, counts = np.unique(subsequences, return_counts=True)    
    probabilities = counts / counts.sum()
    return entropy(probabilities, base=2)

print (f"First {m}x{n} matrix, only ones:")
print(f" - Entropy ignoring order: {round(calculate_entropy(data2d_ones),4)}")
k=2; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d_ones,k),4)}")
k=3; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d_ones,k),4)}")
k=4; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d_ones,k),4)}")

print (f"Second {m}x{n} matrix, random:")
print(f" - Entropy ignoring order: {round(calculate_entropy(data2d_rand),4)}")
k=2; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d_rand,k),4)}")
k=3; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d_rand,k),4)}")
k=4; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d_rand,k),4)}")

print (f"Third {m}x{n} matrix, with pattern:")
print(f" - Entropy ignoring order: {round(calculate_entropy(data2d),4)}")
k=2; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d,k),4)}")
k=3; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d,k),4)}")
k=4; print(f" - Shannon entropy using {k}-bit sequences (k={k}): {round(calculate_sequence_entropy(data2d,k),4)}")
```

### 3. Source Coding

When information is transported over networks, it is essential to compress the amount of data to a minimum. Compression means effectivly to reduce redundancy. The Shannon entropy introduced in the previous section is a known concept in data compression. The entropy of a data source gives a theoretical lower bound on how much the data can be compressed, assuming the most optimal (possibly non-existing) source coding scheme.

#### 3.1 Shannon’s Source Coding Theorem

This theorem states that the average number of bits per symbol (in an optimal encoding) that is needed to represent a message cannot be less than the Shannon entropy of the source.

If a source has an entropy $ H(X) $, then the most you can compress the data (on average) is to use $ H(X) $ bits per symbol. A higher compression cannot be achieved without losing information.

#### 3.2 Compression and Redundancy
A low-entropy source (i.e., with large amount of redundancy) can be compressed more efficiently. For example, the matrix of repeating ones from the previous section has very low entropy and can be compressed significantly. A high-entropy source (i.e., the matrix with random data from the previous section) already carries a lot of information, so there is less redundancy to eliminate, meaning it cannot be compressed much further.

While compression and source coding are related, they serve different purposes in the context of communication networks. In communication theory, source coding is the process of efficiently representing the source information in as few bits as possible. It aims to reduce redundancy from the data source. It is lossless, meaning one can perfectly reconstruct the original data from the encoded data. Compression in general, on the other hand, usually refers to a process where data is encoded into fewer bits to reduce the size for storage or transmission. In some cases (like images or audio), lossy compression is used to achieve higher compression rates, at the expense of some loss in quality. Compression algorithms (e.g., `zlib`, `JPEG`, `MP3`) are used to reduce the size of files or data streams.

#### 3.3 Example Matrices

```python
# --- Data Compression Analysis ---
print (f"First {m}x{n} matrix, only ones:")
bit_string = ''.join(map(str, data2d_ones.flatten()))
compressed_data = zlib.compress(bit_string.encode())
original_size = len(bit_string)
compressed_size = len(compressed_data)
print(f"  - Original size: {original_size} bits")
print(f"  - Compressed size: {compressed_size} bit ({round(compressed_size/original_size*100,2)}% of original size)")

print (f"Second {m}x{n} matrix, random:")
bit_string = ''.join(map(str, data2d_rand.flatten()))
compressed_data = zlib.compress(bit_string.encode())
original_size = len(bit_string)
compressed_size = len(compressed_data)
print(f"  - Original size: {original_size} bits")
print(f"  - Compressed size: {compressed_size} bit ({round(compressed_size/original_size*100,2)}% of original size)")

print (f"Third {m}x{n} matrix, with pattern:")
bit_string = ''.join(map(str, data2d.flatten()))
compressed_data = zlib.compress(bit_string.encode())
original_size = len(bit_string)
compressed_size = len(compressed_data)
print(f"  - Original size: {original_size} bits")
print(f"  - Compressed size: {compressed_size} bits ({round(compressed_size/original_size*100,2)}% of original size)")
```
