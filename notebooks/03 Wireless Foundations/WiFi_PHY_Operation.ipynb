# Wireless Networking and Mobile Computing

### 1 IEEE 802.11 Reference Model

IEEE 802.11 separates functions into the physical layer and the medium access control layer. The physical layer handles modulation, coding, and transmission over the radio channel. The MAC layer handles channel access, framing, acknowledgments, association, and link security. In addition, the MAC Layer Management Entity coordinates management operations such as scanning and authentication. The Station Management Entity spans both layers and provides a control and configuration interface toward higher layers through Service Access Points.

**Layer roles**

- Layer 1: Physical transmission schemes. Modulation, coding, timing, synchronization, channel estimation, OFDM symbol generation

- Layer 2: MAC protocols and MAC Layer Management Entity. Channel access, CSMA CA, interframe spacing, retransmissions, association and authentication

- Layer 1 and 2: Station Management Entity. Configuration, scanning, timing alignment, power management, measurements, key management hooks

- Higher layers: Interfaced through Service Access Points (SAPs)

### 2 Multi-Carrier

#### 2.1 Indoor Radio Channel

Indoor channels are shaped by reflections from walls, furniture, and objects, as well as movement of people. The impulse response therefore contains several delayed components, and the resulting fading varies across both time and frequency. Energy does not arrive along a single path but through many echoes, causing constructive and destructive interference that can change rapidly with small movements. Wideband systems track these variations and adjust modulation, coding, and equalization so the link remains reliable even when individual subcarriers fade or paths shift.

#### 2.1 Multi-Carrier Approach

Orthogonal Frequency Division Multiplexing divides the channel into many orthogonal subcarriers. Each subcarrier carries a narrowband signal, which simplifies equalization and provides robustness against multipath delay spread. The channel is estimated on pilot subcarriers and interpolated across frequency. With OFDMA, the access point allocates groups of subcarriers to different users.

A rectangular grid of narrow subcarriers forms the OFDM spectrum. The cyclic prefix absorbs channel delay so that subcarriers remain orthogonal. As long as frequency offset and timing error are small, carriers do not interfere.

#### 2.2 Multi-Carrier Transmitter

The OFDM transmitter maps bits to QAM symbols, applies coding and interleaving, assigns symbols to subcarriers, inserts pilots, and performs an inverse FFT. A cyclic prefix is added before transmission. The discrete-time representation uses a frequency-domain vector of data and pilot symbols.

Let $X[k]$ be the subcarrier symbols for $k = 0 \ldots N-1$. Then the Time domain OFDM symbol
$$
x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] e^{j2\pi kn / N}
$$

### 4 IEEE 802.11 Spectrum

#### 4.1 The 2.4 GHz Band

The 2.4 GHz band is part of the unlicensed ISM spectrum and is widely used for Wi-Fi. IEEE 802.11 defines a spectral mask that limits energy outside the main channel to avoid excessive interference. Although up to 13 channels exist depending on region, the channels overlap in frequency, and only three are genuinely non-overlapping. Because many unrelated devices operate here, including Bluetooth, cordless phones, and microwave ovens, contention and interference are common.

**Advantages at 2.4 GHz**
- License-free operation and global availability
- Large installed device base and low cost radios

**Disadvantages at 2.4 GHz**
- Only three non-overlapping channels
- Heavily used by heterogeneous devices, leading to congestion and interference

#### 4.2 The 5 GHz Band

The 5 GHz band provides much more spectrum and wider channels for higher throughput. IEEE 802.11a introduced OFDM here with a defined transmit mask that limits out-of-band emissions. Many more non-overlapping channels exist compared to 2.4 GHz, and transmit power limits can be higher depending on the regulatory domain. However, higher frequencies experience more path loss and weaker penetration, which reduces coverage radius and requires denser access point deployment.

**Advantages at 5 GHz**
- More non-overlapping channels and wider bandwidth options
- Often higher allowed transmit power and lower interference levels

**Disadvantages at 5 GHz**
- Higher path loss, reduced indoor range, and weaker wall penetration
- More access points required for uniform coverage in large environments


#### 4.3 Higher vs. Lower Frequency Operation

Lower frequencies such as 2.4 GHz propagate farther and penetrate obstacles more effectively but offer limited spectrum and are crowded. Higher frequencies such as 5 GHz support more channels and higher data rates but suffer greater attenuation and require shorter cell sizes. Real deployments often combine both bands, using 5 GHz for capacity and 2.4 GHz for extended coverage where needed.
