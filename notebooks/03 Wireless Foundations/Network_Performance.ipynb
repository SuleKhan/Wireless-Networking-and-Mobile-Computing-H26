# Wireless Networking and Mobile Computing

### 1 Network Performance Indicators

Network performance cannot be evaluated by throughput alone. Wireless and mobile systems must consider delay, reliability, fairness, energy cost and stability to judge whether a network truly works well. A network that pushes many bits but responds slowly or unfairly may fail in real use. Different applications value different metrics, so a single number cannot capture real performance.

#### 1.1 Key Metrics in Wireless Networking

| Metric         | Meaning                                | Notes                                                         | Typical use relevance            |
| -------------- | -------------------------------------- | ------------------------------------------------------------- | -------------------------------- |
| Throughput     | Data successfully delivered per time   | Measured per link, per stream or per network                  | File transfer, cloud sync        |
| Delay          | Time from sending to receiving         | Includes queueing, processing, retransmission and propagation | Voice, real time control         |
| Jitter         | Variation in delay over time           | Sudden spikes degrade quality                                 | Audio, video, AR VR              |
| Packet loss    | Percentage of lost packets             | Caused by interference, collisions, congestion                | Wireless reliability and quality |
| Fairness       | How evenly resources are shared        | Avoids starving users or flows                                | Multi user networks              |
| Energy per bit | Energy required to send data           | Critical for mobile and IoT devices                           | Sensors, wearables, smartphones  |
| Scalability    | Ability to handle growing device count | Congestion and contention behavior                            | Dense Wi Fi, IoT deployments     |
| Stability      | Consistent performance over time       | Adaptation, control loops, mobility effects                   | Mobile systems, handover         |
