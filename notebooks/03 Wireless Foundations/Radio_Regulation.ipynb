# Wireless Networking and Mobile Computing

### 1 Radio Regulation

The regulation of the radio frequency spectrum is essential for ensuring that different wireless communication systems can operate without interfering with each other. These regulations are managed by international and national agencies, which allocate frequencies, manage interference, and ensure compliance with technical standards.

#### 1.1 Radio Frequency Spectrum

Electromagnetic frequencies between 3 kHz and 300 GHz are commonly referred to as radio frequencies. These frequencies are used for various types of wireless communication, including AM/FM radio, Wi-Fi, cellular, and satellite communications. In vacuum, frequency and wavelength are inversely proportional: the higher the frequency, the shorter the wavelength. The relationship is given by:

$$
\lambda = \frac{c}{f}
$$

where:

* $ \lambda $ = wavelength in meters
* $c$ = speed of light ≈ $ 3.0 \times 10^8 $ m/s
* $f$ = frequency in Hz

#### 1.2 Atmospheric Opacity for Electromagnetic Waves

The Earth's atmosphere affects the propagation of electromagnetic waves by absorbing or blocking certain frequencies. This is due to factors such as water vapor, carbon dioxide, ozone, and the natural absorption at specific wavelengths. As a result, different frequencies experience different levels of attenuation as they pass through the atmosphere.

#### 1.3 Titanic 1912 Tragedy and Radio Regulation

In the 1912 Titanic disaster, distress signals sent via radio were missed due to unreliable radio operations and lack of proper frequency coordination. In response, the U.S. Radio Act was created, establishing international coordination and licensing for radio operators. A dedicated worldwide frequency was allocated for emergency distress calls.

The Radio Act set the framework for:

* Licensing of radio operators and transmissions
* Coordination of frequency usage worldwide
* Spectrum access rights and regulation of communication power, timing, and location

#### 1.4 Regulatory Bodies (Government Agencies)

Several international and national bodies regulate the use of the radio frequency spectrum:

* **International Telecommunications Union (ITU)**: Global coordination and allocation of spectrum
* **United States**:

  * National Telecommunications and Information Administration (NTIA)
  * Federal Communications Commission (FCC)
* **Europe**:

  * European Conference of Postal and Telecommunications Administrations (CEPT)
  * Swiss Federal Office of Communications (OFCOM/BAKOM)

These agencies allocate spectrum, manage interference, and enforce spectrum rights.


### 2 Radio Spectrum Allocation

Radio spectrum allocation is the process of assigning frequencies for specific uses. Some frequencies are reserved for critical applications such as aviation, defense, and emergency communications, while others are allocated for commercial use, like Wi-Fi and cellular. The allocation process ensures that communication systems do not interfere with each other.

#### 2.1 Licensed Radio Spectrum

The majority of the radio spectrum is licensed. This means that commercial license holders are granted exclusive rights to use certain frequencies, which often comes with a licensing fee. Licensed spectrum is typically interference-free, which makes it suitable for critical and high-quality communications, such as cellular networks. Licenses are often long-term (10+ years), but this can lead to inefficiency or unused spectrum.

#### 2.2 Unlicensed Radio Spectrum

Unlicensed spectrum allows anyone to use it without a commercial license, but it is subject to technical and regulatory rules to avoid interference. This spectrum is typically shared by multiple users and can become congested. Systems operating in unlicensed bands must comply with standards to minimize interference. Popular uses include Wi-Fi, Bluetooth, and some IoT technologies.

Unlicensed spectrum has the following limitations:

* Overcrowding can lead to interference from neighboring systems.
* The quality of service cannot be guaranteed by the system due to shared use.

##### FCC Part 15 Rules

The FCC’s Part 15 rules govern the use of unlicensed spectrum in the U.S. Three core principles:

1. **Listen before talk**: Devices must first check for existing transmissions before sending.
2. **Frequent pauses**: Devices should pause often to listen for interference.
3. **Low transmission power**: Devices must avoid overpowering other users.

When a channel is occupied, the device must either wait or move to a different channel.



#### 2.3 Secondary Radio Spectrum Use

* **Ultra Wideband (UWB)**: Utilizes frequencies below the noise floor and operates at very low power.
* **White Space Communication**: Uses unused TV spectrum for short-range wireless communication, typically above the noise floor.
* **Cognitive Radio**: Intelligent systems that can detect available spectrum and adapt their usage dynamically.
