# Wireless Networking and Mobile Computing

### 1 Spread Spectrum Techniques

Wireless systems must operate in environments that include interference, eavesdroppers, multipath fading, and shared frequency resources. Spread spectrum techniques deliberately increase the signal bandwidth beyond what is minimally required for the data rate. While this may seem inefficient, it provides significant gains in robustness, security, multiple access capability, and interference resistance.

##### Historical motivation
Spread spectrum originated from secure military communication ideas in the early 20th century, famously linked to Hedy Lamarr and George Antheil's 1942 patent on frequency hopping concepts for secure torpedo guidance. The core goal was to prevent an adversary from easily jamming or tracking transmissions. Later, Shannon and modern spread-spectrum theory showed that spreading can increase resilience against narrowband interference and improve link reliability in noise-limited and jam-prone environments.

##### General principle
Rather than concentrating energy in a narrow frequency band, the transmitted energy is spread across a wider spectral region in a structured manner. Authorized receivers know how to undo the spreading operation. Uncoordinated receivers perceive the signal as weak, noise-like, or constantly shifting.

A general spreading model is
$$s(t) = m(t) \cdot c(t)$$
where $m(t)$ is the baseband data stream and $c(t)$ is a higher-rate spreading waveform or sequence.

#### 1.1 Frequency Hopping Spread Spectrum

**Definition.**
Frequency Hopping Spread Spectrum (FHSS) rapidly switches the carrier frequency among many narrow channels according to a pseudo-random sequence known to both transmitter and receiver. At any time, the signal occupies only one narrow frequency slot, but over time it spans a wide band. An adversary or interferer must jam all possible hop frequencies simultaneously or predict the hopping pattern. Narrowband interference affects only those hops that overlap with it. Authorized receivers stay synchronized and follow the same hopping schedule.

**Signal model.**
Suppose the hop index at time $t$ is $k(t)$ and the corresponding frequency is $f_{k(t)}$. Then
$$s(t) = m(t)\cos(2\pi f_{k(t)} t)$$
The mapping $k(t)$ follows a pseudo-random sequence known to the receiver. A hop interval $T_h$ defines how long the signal remains on one channel before jumping. Regulatory bodies may define maximum dwell times. Slow hopping holds many data symbols per hop, while fast hopping may hop several times per symbol. To receive successfully, the receiver must align both in time and in the hop sequence. Acquisition takes longer than narrowband systems but once acquired, the link can be highly robust.

**Advantages.**
Frequent hopping mitigates narrowband jammers, bursty interference, and frequency-selective fading. Multi-user FHSS allows multiple devices to hop through channels, colliding only occasionally in time and frequency.

### 1.2 Direct Sequence Spread Spectrum

**Definition.**
Direct Sequence Spread Spectrum (DSSS) multiplies each data symbol by a high-rate pseudo-random chip sequence, increasing the bandwidth proportionally to the chip rate. The receiver uses correlation to reverse the spreading and recover the original data.

**Signal model.**
Let $b[n]$ be the discrete data symbols and $c[k]$ be the chip sequence at a higher rate. The transmitted chip sequence is
$$s[k] = b[n] \cdot c[k] \quad \text{for chips belonging to bit } n$$
This means one bit becomes many chips. The signal bandwidth is determined by the chip rate, not the data rate.

##### Processing gain
$$PG = \frac{R_{\text{chip}}}{R_{\text{data}}}$$
A large processing gain improves the ability to recover signals hidden in noise or shared by many users. After despreading, the desired signal adds coherently, while interference tends to average out. The receiver multiplies the received chips by the same pseudo-random code and integrates over a symbol period. This yields a large value only when the code matches. Timing alignment is crucial. In multipath channels, a rake receiver aligns and combines multiple delayed correlation peaks.

##### Advantages
DSSS supports resistance to narrowband jammers, soft capacity behavior when many coded users share the same band, and immunity to frequency-selective fading when rake combining is used. It also has a “noise-like” appearance which complicates interception without the code.

#### 1.3 Comparing FHSS and DSSS


| Feature                 | FHSS                            | DSSS                                |
| ----------------------- | ------------------------------- | ----------------------------------- |
| Main idea               | Jump across frequencies         | Spread by chip sequences            |
| Instantaneous bandwidth | Narrow per hop                  | Wide continuously                   |
| Interference tolerance  | Very high to narrowband jamming | High, improves with PG              |
| Synchronization         | Sequence and timing             | Chip timing and phase               |
| Multiple access         | Separate hop sets               | Separate spreading codes            |
| Modern use              | Bluetooth Classic, military     | Wi-Fi b, CDMA, GNSS-style receivers |

FHSS is conceptually simpler and tolerant of uncoordinated interference. DSSS achieves higher data rates and stronger processing gain but requires precise synchronization.
