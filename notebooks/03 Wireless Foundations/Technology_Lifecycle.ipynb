# Wireless Networking and Mobile Computing

### 1 Technology

#### 1.1 Technology lifecycle

Wireless technologies move from research ideas to mass-market deployments over years. Some reach broad adoption, some peak in hype and stall, and others slowly mature in niche use cases before scaling. In this course we focus on technologies that have proven value and are being deployed, those currently hyped and rapidly evolving, and those beginning to mature in research and early pilots.

Examples
- Productive markets: 5G, Wi-Fi 6, Bluetooth 5
- Hype phase: satellite broadband, 5G/6G marketing, Wi-Fi 7+
- Early phase: 6G concepts, Li-Fi and visible light networking

#### 1.2 From research to commercialization

Research starts in universities, public programs, and industry labs. Promising ideas move into standardization, where multiple companies align on technical specifications to ensure compatibility. Certification programs confirm interoperability and quality. Spectrum regulation ensures fair and legal operation. Semiconductor and device manufacturers build hardware platforms, and finally operators and consumers adopt and deploy products.

Typical timeline: roughly five to ten years from first research to wide deployment.

Steps
Research
Standardization
Certification
Spectrum regulation
Manufacturing
Deployment

### 2 Standardization

#### 2.1 Why Standardization?
Standardization allows multiple vendors to build compatible systems. This increases product value, reduces vendor lock-in, and shortens development cycles. Key organizations coordinate global wireless standards.

- IEEE: Wi-Fi and many networking standards
- 3GPP: cellular systems (2G to 6G)
- ETSI: European telecom standards and harmonization
- ITU: international coordination, spectrum policy, IMT framework

Standards ensure devices work across countries and networks and support economies of scale.

#### 2.2 Wireless standards evolution: Wi-Fi vs cellular

Wi-Fi and cellular technologies evolved in parallel, each addressing different usage models. Wi-Fi focuses on local connectivity with unlicensed spectrum and cost-efficient deployment. Cellular targets wide-area mobility, licensed spectrum, and QoS under higher mobility. Both families increased throughput, spectral efficiency, and device density over time, and both now adopt MIMO, OFDMA, beamforming, and multi-band operation.

#### 2.4 ITU IMT framework

<center>
  <img src="../../Admin/technology/tech_itu.png" 
       alt="ITU Logo" 
       width="5%"
   >

   International Telecommunication Union (ITU) defines high-level mobile system families under IMT (International Mobile Telecommunications).
</center>


- IMT-2000: 3G era, UMTS, EDGE, DECT
- IMT-Advanced: 4G LTE and related standards, up to ~1 Gb/s
- IMT-2020: 5G family as defined by 3GPP and coordinated globally

The ITU creates global alignment so countries harmonize spectrum and operators achieve roaming compatibility.

#### 2.5 IEEE 802 project

<center>

  IEEE 802 defines standards for LAN and MAN technologies, mainly at OSI layers 1 and 2. It is organized in working groups.

- 802.1 LAN architecture
- 802.3 Ethernet
- 802.11 wireless LAN (Wi-Fi)
- 802.15 wireless personal networks (Bluetooth, UWB)

These standards define the physical layer and link layer behaviors so devices can communicate reliably and interoperably.
